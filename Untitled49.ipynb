```python
import os
import json
import yaml
import joblib
import warnings
import random

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping


# ======================================================================
# STEP 2: RANDOM SEEDS
# ======================================================================

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
random.seed(RANDOM_STATE)
tf.random.set_seed(RANDOM_STATE)


# ======================================================================
# STEP 3: PROJECT PATHS
# ======================================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Rainfall Intensity Classification"
)

DATASET_PATH = os.path.join(
    BASE_DIR,
    "rainfall_0.csv"
)


# ======================================================================
# STEP 4: OUTPUT PATHS
# ======================================================================

H5_PATH = os.path.join(
    BASE_DIR,
    "pso_rainfall_model.h5"
)

PKL_PATH = os.path.join(
    BASE_DIR,
    "pso_rainfall_model.pkl"
)

YAML_PATH = os.path.join(
    BASE_DIR,
    "pso_rainfall_config.yaml"
)

JSON_PATH = os.path.join(
    BASE_DIR,
    "pso_rainfall_metadata.json"
)


ACCURACY_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "pso_accuracy_graph.png"
)

COMPARISON_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "pso_comparison_graph.png"
)

HEATMAP_PATH = os.path.join(
    BASE_DIR,
    "pso_heatmap.png"
)

RESULT_CSV_PATH = os.path.join(
    BASE_DIR,
    "pso_result.csv"
)

RESULT_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "pso_result_graph.png"
)

PREDICTION_CSV_PATH = os.path.join(
    BASE_DIR,
    "pso_prediction.csv"
)

PREDICTION_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "pso_prediction_graph.png"
)

FEATURE_SELECTION_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "pso_feature_selection_graph.png"
)

FEATURE_IMPORTANCE_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "pso_feature_importance_graph.png"
)


os.makedirs(
    BASE_DIR,
    exist_ok=True
)


print("=" * 80)
print("AI-BASED RAINFALL INTENSITY CLASSIFICATION")
print("PARTICLE SWARM OPTIMIZATION (PSO)")
print("=" * 80)


# ======================================================================
# STEP 5: LOAD DATASET
# ======================================================================

if not os.path.exists(DATASET_PATH):

    raise FileNotFoundError(
        f"\nDataset not found:\n{DATASET_PATH}"
    )


df = pd.read_csv(
    DATASET_PATH
)


print("\nDataset loaded successfully.")

print(
    "\nDataset shape:",
    df.shape
)


# ======================================================================
# STEP 6: CLEAN COLUMN NAMES
# ======================================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
)


print("\nDataset columns:")

for column in df.columns:

    print(column)


# ======================================================================
# STEP 7: DETECT LOCATION COLUMNS
# ======================================================================

possible_location_columns = [

    "District",
    "Taluka",
    "Taluk",
    "Circle",
    "Station",
    "Location",
    "Village"

]


location_columns = [

    column

    for column in df.columns

    if column.lower() in [

        value.lower()

        for value in possible_location_columns

    ]

]


print(
    "\nLocation columns:",
    location_columns
)


# ======================================================================
# STEP 8: DETECT RAINFALL COLUMNS
# ======================================================================

candidate_columns = [

    column

    for column in df.columns

    if column not in location_columns

]


rainfall_columns = []


for column in candidate_columns:

    converted = pd.to_numeric(
        df[column],
        errors="coerce"
    )

    numeric_count = (
        converted
        .notna()
        .sum()
    )

    minimum_required = max(
        3,
        int(
            len(df) * 0.30
        )
    )

    if numeric_count >= minimum_required:

        rainfall_columns.append(
            column
        )


if len(rainfall_columns) == 0:

    raise ValueError(
        "No rainfall measurement columns detected."
    )


print(
    "\nNumber of rainfall columns:",
    len(rainfall_columns)
)


print(
    "\nDetected rainfall columns:"
)

for column in rainfall_columns:

    print(column)


# ======================================================================
# STEP 9: CONVERT RAINFALL TO NUMERIC
# ======================================================================

for column in rainfall_columns:

    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )


df[rainfall_columns] = (

    df[rainfall_columns]

    .replace(
        [np.inf, -np.inf],
        np.nan
    )

    .fillna(0)

)


df[rainfall_columns] = (

    df[rainfall_columns]
    .clip(lower=0)

)


print(
    "\nRainfall data cleaned successfully."
)


# ======================================================================
# STEP 10: BASIC FEATURE ENGINEERING
# ======================================================================

rainfall_data = (
    df[rainfall_columns]
)


df["Total_Rainfall"] = (
    rainfall_data
    .sum(axis=1)
)


df["Average_Rainfall"] = (
    rainfall_data
    .mean(axis=1)
)


df["Maximum_Rainfall"] = (
    rainfall_data
    .max(axis=1)
)


df["Minimum_Rainfall"] = (
    rainfall_data
    .min(axis=1)
)


df["Rainfall_Std"] = (
    rainfall_data
    .std(axis=1)
)


df["Rainy_Days"] = (

    rainfall_data > 0

).sum(axis=1)


df["Moderate_Rain_Days"] = (

    rainfall_data >= 15.6

).sum(axis=1)


df["Heavy_Rain_Days"] = (

    rainfall_data >= 64.5

).sum(axis=1)


df["Very_Heavy_Rain_Days"] = (

    rainfall_data >= 115.6

).sum(axis=1)


df["Extreme_Rain_Days"] = (

    rainfall_data >= 204.5

).sum(axis=1)


df["Rainfall_Range"] = (

    df["Maximum_Rainfall"]

    -

    df["Minimum_Rainfall"]

)


df["Rainfall_CV"] = (

    df["Rainfall_Std"]

    /

    (
        df["Average_Rainfall"]
        +
        1e-6
    )

)


# ======================================================================
# STEP 11: MONTH FEATURES
# ======================================================================

august_columns = [

    column

    for column in rainfall_columns

    if "aug" in column.lower()

]


september_columns = [

    column

    for column in rainfall_columns

    if "sep" in column.lower()

]


if len(august_columns) > 0:

    df["August_Total_Rainfall"] = (

        df[august_columns]
        .sum(axis=1)

    )

    df["August_Average_Rainfall"] = (

        df[august_columns]
        .mean(axis=1)

    )

else:

    df["August_Total_Rainfall"] = 0

    df["August_Average_Rainfall"] = 0


if len(september_columns) > 0:

    df["September_Total_Rainfall"] = (

        df[september_columns]
        .sum(axis=1)

    )

    df["September_Average_Rainfall"] = (

        df[september_columns]
        .mean(axis=1)

    )

else:

    df["September_Total_Rainfall"] = 0

    df["September_Average_Rainfall"] = 0


print(
    "\nFeature engineering completed."
)


# ======================================================================
# STEP 12: CREATE RAINFALL INTENSITY TARGET
# ======================================================================

lower_threshold = (

    df["Total_Rainfall"]
    .quantile(0.33)

)


upper_threshold = (

    df["Total_Rainfall"]
    .quantile(0.66)

)


def classify_rainfall(value):

    if value <= lower_threshold:

        return "Low"

    elif value <= upper_threshold:

        return "Moderate"

    else:

        return "High"


df["Rainfall_Intensity"] = (

    df["Total_Rainfall"]
    .apply(
        classify_rainfall
    )

)


print(
    "\nRainfall classification thresholds:"
)

print(
    "Low <=",
    round(
        lower_threshold,
        2
    )
)

print(
    "Moderate <=",
    round(
        upper_threshold,
        2
    )
)

print(
    "High >",
    round(
        upper_threshold,
        2
    )
)


print(
    "\nClass Distribution:"
)

print(
    df["Rainfall_Intensity"]
    .value_counts()
)


# ======================================================================
# STEP 13: DEFINE CANDIDATE FEATURES
# ======================================================================

all_features = [

    "Average_Rainfall",

    "Maximum_Rainfall",

    "Minimum_Rainfall",

    "Rainfall_Std",

    "Rainy_Days",

    "Moderate_Rain_Days",

    "Heavy_Rain_Days",

    "Very_Heavy_Rain_Days",

    "Extreme_Rain_Days",

    "Rainfall_Range",

    "Rainfall_CV",

    "August_Total_Rainfall",

    "August_Average_Rainfall",

    "September_Total_Rainfall",

    "September_Average_Rainfall"

]


X = (

    df[
        all_features
    ]

    .copy()

)


X = (

    X.replace(
        [np.inf, -np.inf],
        np.nan
    )

    .fillna(0)

)


y = (

    df[
        "Rainfall_Intensity"
    ]

    .copy()

)


print(
    "\nCandidate features:",
    len(all_features)
)


# ======================================================================
# STEP 14: LABEL ENCODING
# ======================================================================

label_encoder = (
    LabelEncoder()
)


y_encoded = (

    label_encoder
    .fit_transform(
        y
    )

)


print(
    "\nEncoded classes:"
)


for index, label in enumerate(
    label_encoder.classes_
):

    print(
        index,
        "=",
        label
    )


# ======================================================================
# STEP 15: TRAIN TEST SPLIT
# ======================================================================

indices = np.arange(
    len(df)
)


(
    X_train_full,
    X_test_full,
    y_train,
    y_test,
    train_indices,
    test_indices

) = train_test_split(

    X,

    y_encoded,

    indices,

    test_size=0.25,

    random_state=RANDOM_STATE,

    stratify=y_encoded

)


print(
    "\nTraining samples:",
    len(X_train_full)
)


print(
    "Testing samples:",
    len(X_test_full)
)


# ======================================================================
# STEP 16: CREATE PSO TRAINING AND VALIDATION DATA
# ======================================================================

try:

    (
        X_pso_train,
        X_pso_validation,
        y_pso_train,
        y_pso_validation

    ) = train_test_split(

        X_train_full,

        y_train,

        test_size=0.25,

        random_state=RANDOM_STATE,

        stratify=y_train

    )

except ValueError:

    (
        X_pso_train,
        X_pso_validation,
        y_pso_train,
        y_pso_validation

    ) = train_test_split(

        X_train_full,

        y_train,

        test_size=0.25,

        random_state=RANDOM_STATE

    )


print(
    "\nPSO training samples:",
    len(X_pso_train)
)


print(
    "PSO validation samples:",
    len(X_pso_validation)
)


# ======================================================================
# STEP 17: PSO CONFIGURATION
# ======================================================================

NUMBER_OF_FEATURES = len(
    all_features
)

NUMBER_OF_PARTICLES = 20

NUMBER_OF_ITERATIONS = 25

INERTIA_WEIGHT = 0.729

COGNITIVE_CONSTANT = 1.49445

SOCIAL_CONSTANT = 1.49445

FEATURE_PENALTY = 0.01


print(
    "\nPSO Configuration:"
)

print(
    "Particles:",
    NUMBER_OF_PARTICLES
)

print(
    "Iterations:",
    NUMBER_OF_ITERATIONS
)

print(
    "Inertia:",
    INERTIA_WEIGHT
)

print(
    "Cognitive constant:",
    COGNITIVE_CONSTANT
)

print(
    "Social constant:",
    SOCIAL_CONSTANT
)


# ======================================================================
# STEP 18: PSO FITNESS FUNCTION
# ======================================================================

def calculate_fitness(position):

    selected_indices = np.where(
        position == 1
    )[0]


    if len(selected_indices) == 0:

        return 0.0


    selected_features = [

        all_features[index]

        for index in selected_indices

    ]


    X_train_selected = (

        X_pso_train[
            selected_features
        ]

    )


    X_validation_selected = (

        X_pso_validation[
            selected_features
        ]

    )


    model = RandomForestClassifier(

        n_estimators=120,

        class_weight="balanced",

        random_state=RANDOM_STATE,

        n_jobs=-1

    )


    model.fit(

        X_train_selected,

        y_pso_train

    )


    prediction = model.predict(

        X_validation_selected

    )


    accuracy = accuracy_score(

        y_pso_validation,

        prediction

    )


    feature_ratio = (

        len(selected_features)

        /

        NUMBER_OF_FEATURES

    )


    fitness = (

        accuracy

        -

        (
            FEATURE_PENALTY
            *
            feature_ratio
        )

    )


    return fitness


# ======================================================================
# STEP 19: SIGMOID FUNCTION
# ======================================================================

def sigmoid(value):

    value = np.clip(
        value,
        -20,
        20
    )

    return (

        1.0

        /

        (
            1.0
            +
            np.exp(
                -value
            )
        )

    )


# ======================================================================
# STEP 20: INITIALIZE PSO PARTICLES
# ======================================================================

positions = np.random.randint(

    0,

    2,

    size=(

        NUMBER_OF_PARTICLES,

        NUMBER_OF_FEATURES

    )

)


# Ensure each particle has at least one feature
for particle_index in range(
    NUMBER_OF_PARTICLES
):

    if np.sum(
        positions[
            particle_index
        ]
    ) == 0:

        random_feature = np.random.randint(

            0,

            NUMBER_OF_FEATURES

        )

        positions[
            particle_index,
            random_feature
        ] = 1


velocities = np.random.uniform(

    low=-1,

    high=1,

    size=(

        NUMBER_OF_PARTICLES,

        NUMBER_OF_FEATURES

    )

)


# ======================================================================
# STEP 21: PERSONAL BEST INITIALIZATION
# ======================================================================

personal_best_positions = (
    positions.copy()
)


personal_best_scores = np.full(

    NUMBER_OF_PARTICLES,

    -np.inf

)


global_best_position = None

global_best_score = -np.inf


# ======================================================================
# STEP 22: PSO HISTORY
# ======================================================================

pso_history = []

pso_feature_count_history = []


# ======================================================================
# STEP 23: RUN BINARY PSO
# ======================================================================

print(
    "\n"
    +
    "=" * 80
)

print(
    "STARTING BINARY PARTICLE SWARM OPTIMIZATION"
)

print(
    "=" * 80
)


for iteration in range(
    NUMBER_OF_ITERATIONS
):


    # ------------------------------------------------------------------
    # Evaluate particles
    # ------------------------------------------------------------------

    for particle_index in range(
        NUMBER_OF_PARTICLES
    ):


        current_position = (

            positions[
                particle_index
            ]

        )


        current_fitness = calculate_fitness(

            current_position

        )


        # --------------------------------------------------------------
        # Update personal best
        # --------------------------------------------------------------

        if current_fitness > personal_best_scores[
            particle_index
        ]:

            personal_best_scores[
                particle_index
            ] = current_fitness

            personal_best_positions[
                particle_index
            ] = current_position.copy()


        # --------------------------------------------------------------
        # Update global best
        # --------------------------------------------------------------

        if current_fitness > global_best_score:

            global_best_score = (
                current_fitness
            )

            global_best_position = (
                current_position.copy()
            )


    # ------------------------------------------------------------------
    # Store history
    # ------------------------------------------------------------------

    pso_history.append(
        global_best_score
    )


    selected_feature_count = int(

        np.sum(
            global_best_position
        )

    )


    pso_feature_count_history.append(
        selected_feature_count
    )


    print(

        f"Iteration "
        f"{iteration + 1:02d}/"
        f"{NUMBER_OF_ITERATIONS}"

        f" | Best Fitness: "
        f"{global_best_score:.4f}"

        f" | Selected Features: "
        f"{selected_feature_count}"

    )


    # ------------------------------------------------------------------
    # Update velocity and position
    # ------------------------------------------------------------------

    for particle_index in range(
        NUMBER_OF_PARTICLES
    ):


        r1 = np.random.random(
            NUMBER_OF_FEATURES
        )


        r2 = np.random.random(
            NUMBER_OF_FEATURES
        )


        cognitive_component = (

            COGNITIVE_CONSTANT

            *

            r1

            *

            (
                personal_best_positions[
                    particle_index
                ]

                -

                positions[
                    particle_index
                ]
            )

        )


        social_component = (

            SOCIAL_CONSTANT

            *

            r2

            *

            (
                global_best_position

                -

                positions[
                    particle_index
                ]
            )

        )


        velocities[
            particle_index
        ] = (

            INERTIA_WEIGHT

            *

            velocities[
                particle_index
            ]

            +

            cognitive_component

            +

            social_component

        )


        probabilities = sigmoid(

            velocities[
                particle_index
            ]

        )


        random_values = np.random.random(

            NUMBER_OF_FEATURES

        )


        positions[
            particle_index
        ] = (

            random_values

            <

            probabilities

        ).astype(int)


        # --------------------------------------------------------------
        # Ensure particle selects at least one feature
        # --------------------------------------------------------------

        if np.sum(
            positions[
                particle_index
            ]
        ) == 0:

            random_feature = np.random.randint(

                0,

                NUMBER_OF_FEATURES

            )

            positions[
                particle_index,
                random_feature
            ] = 1


# ======================================================================
# STEP 24: GET BEST PSO FEATURES
# ======================================================================

selected_indices = np.where(

    global_best_position == 1

)[0]


selected_features = [

    all_features[index]

    for index in selected_indices

]


print(
    "\n"
    +
    "=" * 80
)

print(
    "PSO FEATURE SELECTION COMPLETED"
)

print(
    "=" * 80
)


print(
    "\nBest PSO fitness:",
    round(
        global_best_score,
        4
    )
)


print(
    "\nNumber of selected features:",
    len(selected_features)
)


print(
    "\nPSO selected features:"
)


for number, feature in enumerate(

    selected_features,

    start=1

):

    print(
        f"{number}. {feature}"
    )


# ======================================================================
# STEP 25: PSO OPTIMIZATION GRAPH
# ======================================================================

plt.figure(
    figsize=(10, 6)
)


plt.plot(

    range(
        1,
        len(pso_history) + 1
    ),

    pso_history,

    marker="o",

    linewidth=2

)


plt.title(

    "PSO Feature Selection Optimization",

    fontsize=14,

    fontweight="bold"

)


plt.xlabel(
    "Iteration"
)


plt.ylabel(
    "Global Best Fitness"
)


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    FEATURE_SELECTION_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 26: PREPARE PSO SELECTED FEATURES
# ======================================================================

X_selected = (

    X[
        selected_features
    ]

    .copy()

)


X_train = (

    X_train_full[
        selected_features
    ]

    .copy()

)


X_test = (

    X_test_full[
        selected_features
    ]

    .copy()

)


print(
    "\nFinal PSO training shape:",
    X_train.shape
)


print(
    "Final PSO testing shape:",
    X_test.shape
)


# ======================================================================
# STEP 27: STANDARD SCALER
# ======================================================================

scaler = (
    StandardScaler()
)


X_train_scaled = scaler.fit_transform(

    X_train

)


X_test_scaled = scaler.transform(

    X_test

)


# ======================================================================
# STEP 28: TRAIN PSO RANDOM FOREST
# ======================================================================

print(
    "\nTraining PSO-optimized Random Forest..."
)


pso_rf_model = RandomForestClassifier(

    n_estimators=500,

    class_weight="balanced",

    random_state=RANDOM_STATE,

    n_jobs=-1

)


pso_rf_model.fit(

    X_train,

    y_train

)


# ======================================================================
# STEP 29: RANDOM FOREST PREDICTION
# ======================================================================

pso_rf_prediction = (

    pso_rf_model
    .predict(
        X_test
    )

)


# ======================================================================
# STEP 30: RANDOM FOREST METRICS
# ======================================================================

pso_rf_accuracy = accuracy_score(

    y_test,

    pso_rf_prediction

)


pso_rf_precision = precision_score(

    y_test,

    pso_rf_prediction,

    average="weighted",

    zero_division=0

)


pso_rf_recall = recall_score(

    y_test,

    pso_rf_prediction,

    average="weighted",

    zero_division=0

)


pso_rf_f1 = f1_score(

    y_test,

    pso_rf_prediction,

    average="weighted",

    zero_division=0

)


print(
    "\nPSO Random Forest Accuracy:",
    f"{pso_rf_accuracy * 100:.2f}%"
)


# ======================================================================
# STEP 31: SAVE PSO RANDOM FOREST PKL
# ======================================================================

pso_model_package = {

    "model":
        pso_rf_model,

    "scaler":
        scaler,

    "label_encoder":
        label_encoder,

    "selected_features":
        selected_features,

    "all_features":
        all_features,

    "best_particle":
        global_best_position,

    "best_fitness":
        global_best_score,

    "rainfall_columns":
        rainfall_columns,

    "lower_threshold":
        lower_threshold,

    "upper_threshold":
        upper_threshold

}


joblib.dump(

    pso_model_package,

    PKL_PATH

)


print(
    "\nPSO PKL saved:"
)

print(
    PKL_PATH
)


# ======================================================================
# STEP 32: PREPARE NEURAL NETWORK TARGET
# ======================================================================

number_of_classes = len(

    label_encoder.classes_

)


y_train_nn = (

    tf.keras.utils
    .to_categorical(

        y_train,

        num_classes=
        number_of_classes

    )

)


y_test_nn = (

    tf.keras.utils
    .to_categorical(

        y_test,

        num_classes=
        number_of_classes

    )

)


# ======================================================================
# STEP 33: BUILD PSO NEURAL NETWORK
# ======================================================================

print(
    "\nTraining PSO-optimized Neural Network..."
)


pso_nn_model = Sequential([

    tf.keras.layers.Input(

        shape=(
            len(selected_features),
        )

    ),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(
        0.20
    ),

    Dense(
        32,
        activation="relu"
    ),

    Dropout(
        0.15
    ),

    Dense(
        16,
        activation="relu"
    ),

    Dense(

        number_of_classes,

        activation="softmax"

    )

])


# ======================================================================
# STEP 34: COMPILE NEURAL NETWORK
# ======================================================================

pso_nn_model.compile(

    optimizer=tf.keras.optimizers.Adam(

        learning_rate=0.001

    ),

    loss="categorical_crossentropy",

    metrics=[
        "accuracy"
    ]

)


pso_nn_model.summary()


# ======================================================================
# STEP 35: EARLY STOPPING
# ======================================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=20,

    restore_best_weights=True,

    verbose=1

)


# ======================================================================
# STEP 36: TRAIN PSO NEURAL NETWORK
# ======================================================================

history = pso_nn_model.fit(

    X_train_scaled,

    y_train_nn,

    validation_split=0.20,

    epochs=200,

    batch_size=8,

    callbacks=[
        early_stopping
    ],

    verbose=1

)


# ======================================================================
# STEP 37: PSO NEURAL NETWORK PREDICTION
# ======================================================================

pso_nn_probability = (

    pso_nn_model
    .predict(

        X_test_scaled,

        verbose=0

    )

)


pso_nn_prediction = np.argmax(

    pso_nn_probability,

    axis=1

)


# ======================================================================
# STEP 38: NEURAL NETWORK METRICS
# ======================================================================

pso_nn_accuracy = accuracy_score(

    y_test,

    pso_nn_prediction

)


pso_nn_precision = precision_score(

    y_test,

    pso_nn_prediction,

    average="weighted",

    zero_division=0

)


pso_nn_recall = recall_score(

    y_test,

    pso_nn_prediction,

    average="weighted",

    zero_division=0

)


pso_nn_f1 = f1_score(

    y_test,

    pso_nn_prediction,

    average="weighted",

    zero_division=0

)


print(
    "\nPSO Neural Network Accuracy:",
    f"{pso_nn_accuracy * 100:.2f}%"
)


# ======================================================================
# STEP 39: SAVE PSO H5 MODEL
# ======================================================================

pso_nn_model.save(

    H5_PATH

)


print(
    "\nPSO H5 saved:"
)

print(
    H5_PATH
)


# ======================================================================
# STEP 40: PSO ACCURACY GRAPH
# ======================================================================

epochs = range(

    1,

    len(
        history.history[
            "accuracy"
        ]
    )
    +
    1

)


plt.figure(
    figsize=(10, 6)
)


plt.plot(

    epochs,

    history.history[
        "accuracy"
    ],

    marker="o",

    markersize=3,

    label="Training Accuracy"

)


plt.plot(

    epochs,

    history.history[
        "val_accuracy"
    ],

    marker="s",

    markersize=3,

    label="Validation Accuracy"

)


plt.title(

    "PSO Neural Network Training and Validation Accuracy",

    fontsize=14,

    fontweight="bold"

)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "Accuracy"
)


plt.ylim(
    0,
    1.05
)


plt.grid(
    alpha=0.3
)


plt.legend()


plt.tight_layout()


plt.savefig(

    ACCURACY_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 41: MODEL COMPARISON DATA
# ======================================================================

comparison_data = pd.DataFrame({

    "Model": [

        "PSO Random Forest",

        "PSO Neural Network"

    ],

    "Accuracy": [

        pso_rf_accuracy,

        pso_nn_accuracy

    ],

    "Precision": [

        pso_rf_precision,

        pso_nn_precision

    ],

    "Recall": [

        pso_rf_recall,

        pso_nn_recall

    ],

    "F1 Score": [

        pso_rf_f1,

        pso_nn_f1

    ]

})


print(
    "\nPSO Model Comparison:"
)


print(
    comparison_data
    .to_string(
        index=False
    )
)


# ======================================================================
# STEP 42: PSO COMPARISON GRAPH
# ======================================================================

comparison_plot = (

    comparison_data

    .set_index(
        "Model"
    )

    *

    100

)


ax = comparison_plot.plot(

    kind="bar",

    figsize=(11, 6),

    width=0.75

)


plt.title(

    "PSO-Optimized Model Performance Comparison",

    fontsize=14,

    fontweight="bold"

)


plt.xlabel(
    "Model"
)


plt.ylabel(
    "Score (%)"
)


plt.ylim(
    0,
    105
)


plt.xticks(
    rotation=0
)


plt.grid(

    axis="y",

    alpha=0.3

)


plt.legend(
    title="Metrics"
)


for container in ax.containers:

    ax.bar_label(

        container,

        fmt="%.1f",

        padding=3,

        fontsize=8

    )


plt.tight_layout()


plt.savefig(

    COMPARISON_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 43: PSO CONFUSION MATRIX / HEATMAP
# ======================================================================

cm = confusion_matrix(

    y_test,

    pso_rf_prediction

)


fig, ax = plt.subplots(

    figsize=(8, 6)

)


image = ax.imshow(

    cm,

    interpolation="nearest",

    cmap="Blues"

)


plt.colorbar(
    image
)


classes = (
    label_encoder.classes_
)


ax.set(

    xticks=np.arange(
        len(classes)
    ),

    yticks=np.arange(
        len(classes)
    ),

    xticklabels=classes,

    yticklabels=classes,

    xlabel="Predicted Rainfall Intensity",

    ylabel="Actual Rainfall Intensity",

    title="PSO Random Forest Confusion Matrix"

)


threshold = (

    cm.max() / 2

    if cm.size > 0

    else 0

)


for i in range(
    cm.shape[0]
):

    for j in range(
        cm.shape[1]
    ):

        ax.text(

            j,

            i,

            str(
                cm[i, j]
            ),

            ha="center",

            va="center",

            color=(

                "white"

                if cm[i, j] > threshold

                else "black"

            ),

            fontsize=12,

            fontweight="bold"

        )


plt.tight_layout()


plt.savefig(

    HEATMAP_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 44: PSO FEATURE IMPORTANCE
# ======================================================================

feature_importance = pd.DataFrame({

    "Feature":
        selected_features,

    "Importance":
        pso_rf_model.feature_importances_

})


feature_importance = (

    feature_importance

    .sort_values(

        "Importance",

        ascending=True

    )

)


plt.figure(
    figsize=(10, 7)
)


plt.barh(

    feature_importance[
        "Feature"
    ],

    feature_importance[
        "Importance"
    ]

)


plt.title(

    "PSO Selected Feature Importance",

    fontsize=14,

    fontweight="bold"

)


plt.xlabel(
    "Importance Score"
)


plt.ylabel(
    "Selected Feature"
)


plt.grid(

    axis="x",

    alpha=0.3

)


plt.tight_layout()


plt.savefig(

    FEATURE_IMPORTANCE_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 45: CREATE PSO RESULT CSV
# ======================================================================

actual_labels = (

    label_encoder
    .inverse_transform(
        y_test
    )

)


rf_labels = (

    label_encoder
    .inverse_transform(
        pso_rf_prediction
    )

)


nn_labels = (

    label_encoder
    .inverse_transform(
        pso_nn_prediction
    )

)


test_df = (

    df.iloc[
        test_indices
    ]

    .copy()

    .reset_index(
        drop=True
    )

)


result_output = (
    pd.DataFrame()
)


for column in location_columns:

    result_output[
        column
    ] = test_df[
        column
    ]


result_output[
    "Total_Rainfall"
] = test_df[
    "Total_Rainfall"
]


result_output[
    "Average_Rainfall"
] = test_df[
    "Average_Rainfall"
]


result_output[
    "Maximum_Rainfall"
] = test_df[
    "Maximum_Rainfall"
]


result_output[
    "Rainy_Days"
] = test_df[
    "Rainy_Days"
]


result_output[
    "Heavy_Rain_Days"
] = test_df[
    "Heavy_Rain_Days"
]


result_output[
    "Actual_Intensity"
] = actual_labels


result_output[
    "PSO_RF_Prediction"
] = rf_labels


result_output[
    "PSO_NN_Prediction"
] = nn_labels


result_output[
    "PSO_RF_Correct"
] = (

    result_output[
        "Actual_Intensity"
    ]

    ==

    result_output[
        "PSO_RF_Prediction"
    ]

)


result_output[
    "PSO_NN_Correct"
] = (

    result_output[
        "Actual_Intensity"
    ]

    ==

    result_output[
        "PSO_NN_Prediction"
    ]

)


result_output.to_csv(

    RESULT_CSV_PATH,

    index=False

)


print(
    "\nPSO result CSV saved:"
)

print(
    RESULT_CSV_PATH
)


print(
    "\nPSO Result Preview:"
)

print(
    result_output.head(10)
)


# ======================================================================
# STEP 46: PSO RESULT GRAPH
# ======================================================================

actual_counts = (

    result_output[
        "Actual_Intensity"
    ]

    .value_counts()

    .reindex(
        classes,
        fill_value=0
    )

)


rf_counts = (

    result_output[
        "PSO_RF_Prediction"
    ]

    .value_counts()

    .reindex(
        classes,
        fill_value=0
    )

)


nn_counts = (

    result_output[
        "PSO_NN_Prediction"
    ]

    .value_counts()

    .reindex(
        classes,
        fill_value=0
    )

)


x_position = np.arange(
    len(classes)
)


width = 0.25


plt.figure(
    figsize=(10, 6)
)


plt.bar(

    x_position - width,

    actual_counts.values,

    width,

    label="Actual"

)


plt.bar(

    x_position,

    rf_counts.values,

    width,

    label="PSO Random Forest"

)


plt.bar(

    x_position + width,

    nn_counts.values,

    width,

    label="PSO Neural Network"

)


plt.title(

    "PSO Actual vs Predicted Rainfall Intensity",

    fontsize=14,

    fontweight="bold"

)


plt.xlabel(
    "Rainfall Intensity"
)


plt.ylabel(
    "Number of Records"
)


plt.xticks(

    x_position,

    classes

)


plt.legend()


plt.grid(

    axis="y",

    alpha=0.3

)


plt.tight_layout()


plt.savefig(

    RESULT_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 47: PREDICT COMPLETE DATASET
# ======================================================================

X_all_selected = (

    X[
        selected_features
    ]

    .copy()

)


# ----------------------------------------------------------------------
# RANDOM FOREST
# ----------------------------------------------------------------------

pso_rf_all_prediction = (

    pso_rf_model
    .predict(
        X_all_selected
    )

)


pso_rf_all_probability = (

    pso_rf_model
    .predict_proba(
        X_all_selected
    )

)


pso_rf_all_labels = (

    label_encoder
    .inverse_transform(
        pso_rf_all_prediction
    )

)


# ----------------------------------------------------------------------
# NEURAL NETWORK
# ----------------------------------------------------------------------

X_all_scaled = (

    scaler
    .transform(
        X_all_selected
    )

)


pso_nn_all_probability = (

    pso_nn_model
    .predict(

        X_all_scaled,

        verbose=0

    )

)


pso_nn_all_prediction = np.argmax(

    pso_nn_all_probability,

    axis=1

)


pso_nn_all_labels = (

    label_encoder
    .inverse_transform(
        pso_nn_all_prediction
    )

)


# ======================================================================
# STEP 48: CREATE PSO PREDICTION CSV
# ======================================================================

prediction_output = (
    pd.DataFrame()
)


for column in location_columns:

    prediction_output[
        column
    ] = df[
        column
    ]


prediction_output[
    "Total_Rainfall"
] = df[
    "Total_Rainfall"
]


prediction_output[
    "Average_Rainfall"
] = df[
    "Average_Rainfall"
]


prediction_output[
    "Maximum_Rainfall"
] = df[
    "Maximum_Rainfall"
]


prediction_output[
    "Rainfall_Std"
] = df[
    "Rainfall_Std"
]


prediction_output[
    "Rainy_Days"
] = df[
    "Rainy_Days"
]


prediction_output[
    "Heavy_Rain_Days"
] = df[
    "Heavy_Rain_Days"
]


prediction_output[
    "Actual_Intensity"
] = df[
    "Rainfall_Intensity"
]


prediction_output[
    "PSO_RF_Prediction"
] = (
    pso_rf_all_labels
)


prediction_output[
    "PSO_NN_Prediction"
] = (
    pso_nn_all_labels
)


prediction_output[
    "PSO_RF_Confidence"
] = (

    np.max(

        pso_rf_all_probability,

        axis=1

    )

    *

    100

)


prediction_output[
    "PSO_NN_Confidence"
] = (

    np.max(

        pso_nn_all_probability,

        axis=1

    )

    *

    100

)


prediction_output.to_csv(

    PREDICTION_CSV_PATH,

    index=False

)


print(
    "\nPSO prediction CSV saved:"
)

print(
    PREDICTION_CSV_PATH
)


print(
    "\nPSO Prediction Preview:"
)

print(
    prediction_output.head(10)
)


# ======================================================================
# STEP 49: PSO PREDICTION GRAPH
# ======================================================================

prediction_counts = (

    prediction_output[
        "PSO_RF_Prediction"
    ]

    .value_counts()

    .reindex(
        classes,
        fill_value=0
    )

)


plt.figure(
    figsize=(9, 6)
)


bars = plt.bar(

    prediction_counts.index,

    prediction_counts.values

)


plt.title(

    "PSO Predicted Rainfall Intensity Distribution",

    fontsize=14,

    fontweight="bold"

)


plt.xlabel(
    "Rainfall Intensity"
)


plt.ylabel(
    "Number of Locations"
)


plt.grid(

    axis="y",

    alpha=0.3

)


for bar in bars:

    height = (
        bar.get_height()
    )

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        height,

        str(
            int(height)
        ),

        ha="center",

        va="bottom",

        fontweight="bold"

    )


plt.tight_layout()


plt.savefig(

    PREDICTION_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 50: SAVE PSO YAML
# ======================================================================

yaml_config = {

    "project": {

        "name":
            "AI-Based Rainfall Intensity Classification",

        "optimization":
            "Particle Swarm Optimization",

        "pso_method":
            "Binary PSO Feature Selection"

    },


    "dataset": {

        "filename":
            os.path.basename(
                DATASET_PATH
            ),

        "rows":
            int(
                df.shape[0]
            ),

        "rainfall_columns":
            int(
                len(
                    rainfall_columns
                )
            )

    },


    "pso": {

        "particles":
            NUMBER_OF_PARTICLES,

        "iterations":
            NUMBER_OF_ITERATIONS,

        "inertia_weight":
            INERTIA_WEIGHT,

        "cognitive_constant":
            COGNITIVE_CONSTANT,

        "social_constant":
            SOCIAL_CONSTANT,

        "feature_penalty":
            FEATURE_PENALTY,

        "best_fitness":
            float(
                global_best_score
            ),

        "selected_feature_count":
            len(
                selected_features
            ),

        "selected_features":
            selected_features

    },


    "target": {

        "column":
            "Rainfall_Intensity",

        "classes":
            label_encoder
            .classes_
            .tolist()

    },


    "models": {

        "pso_random_forest": {

            "accuracy":
                float(
                    pso_rf_accuracy
                ),

            "precision":
                float(
                    pso_rf_precision
                ),

            "recall":
                float(
                    pso_rf_recall
                ),

            "f1_score":
                float(
                    pso_rf_f1
                )

        },


        "pso_neural_network": {

            "accuracy":
                float(
                    pso_nn_accuracy
                ),

            "precision":
                float(
                    pso_nn_precision
                ),

            "recall":
                float(
                    pso_nn_recall
                ),

            "f1_score":
                float(
                    pso_nn_f1
                )

        }

    }

}


with open(

    YAML_PATH,

    "w",

    encoding="utf-8"

) as file:

    yaml.safe_dump(

        yaml_config,

        file,

        sort_keys=False,

        allow_unicode=True

    )


print(
    "\nPSO YAML saved:"
)

print(
    YAML_PATH
)


# ======================================================================
# STEP 51: SAVE PSO JSON
# ======================================================================

json_metadata = {

    "project_name":
        "AI-Based Rainfall Intensity Classification",

    "optimization_algorithm":
        "Particle Swarm Optimization",

    "pso_implementation":
        "Binary PSO Feature Selection",

    "dataset":
        os.path.basename(
            DATASET_PATH
        ),

    "dataset_rows":
        int(
            len(df)
        ),

    "candidate_features":
        all_features,

    "selected_features":
        selected_features,

    "selected_feature_count":
        int(
            len(
                selected_features
            )
        ),

    "best_particle":
        global_best_position
        .astype(int)
        .tolist(),

    "best_fitness":
        float(
            global_best_score
        ),


    "pso_parameters": {

        "number_of_particles":
            NUMBER_OF_PARTICLES,

        "number_of_iterations":
            NUMBER_OF_ITERATIONS,

        "inertia_weight":
            INERTIA_WEIGHT,

        "cognitive_constant":
            COGNITIVE_CONSTANT,

        "social_constant":
            SOCIAL_CONSTANT,

        "feature_penalty":
            FEATURE_PENALTY

    },


    "rainfall_thresholds": {

        "low_upper":
            float(
                lower_threshold
            ),

        "moderate_upper":
            float(
                upper_threshold
            )

    },


    "pso_random_forest": {

        "accuracy":
            float(
                pso_rf_accuracy
            ),

        "precision":
            float(
                pso_rf_precision
            ),

        "recall":
            float(
                pso_rf_recall
            ),

        "f1_score":
            float(
                pso_rf_f1
            )

    },


    "pso_neural_network": {

        "accuracy":
            float(
                pso_nn_accuracy
            ),

        "precision":
            float(
                pso_nn_precision
            ),

        "recall":
            float(
                pso_nn_recall
            ),

        "f1_score":
            float(
                pso_nn_f1
            )

    }

}


with open(

    JSON_PATH,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        json_metadata,

        file,

        indent=4

    )


print(
    "\nPSO JSON saved:"
)

print(
    JSON_PATH
)


# ======================================================================
# STEP 52: RANDOM FOREST CLASSIFICATION REPORT
# ======================================================================

print(
    "\n"
    +
    "=" * 80
)

print(
    "PSO RANDOM FOREST CLASSIFICATION REPORT"
)

print(
    "=" * 80
)


print(

    classification_report(

        y_test,

        pso_rf_prediction,

        target_names=
        label_encoder.classes_,

        zero_division=0

    )

)


# ======================================================================
# STEP 53: NEURAL NETWORK CLASSIFICATION REPORT
# ======================================================================

print(
    "\n"
    +
    "=" * 80
)

print(
    "PSO NEURAL NETWORK CLASSIFICATION REPORT"
)

print(
    "=" * 80
)


print(

    classification_report(

        y_test,

        pso_nn_prediction,

        target_names=
        label_encoder.classes_,

        zero_division=0

    )

)


# ======================================================================
# STEP 54: TEST SAVED PSO PKL MODEL
# ======================================================================

print(
    "\nTesting saved PSO PKL model..."
)


loaded_package = joblib.load(
    PKL_PATH
)


loaded_rf_model = (
    loaded_package[
        "model"
    ]
)


sample = (

    X_selected
    .iloc[
        [0]
    ]

)


sample_prediction = (

    loaded_rf_model
    .predict(
        sample
    )

)


sample_label = (

    label_encoder
    .inverse_transform(
        sample_prediction
    )[0]

)


print(
    "PSO PKL sample prediction:",
    sample_label
)


# ======================================================================
# STEP 55: TEST SAVED PSO H5 MODEL
# ======================================================================

print(
    "\nTesting saved PSO H5 model..."
)


loaded_nn_model = (

    tf.keras.models
    .load_model(
        H5_PATH
    )

)


sample_scaled = (

    scaler
    .transform(
        sample
    )

)


sample_nn_probability = (

    loaded_nn_model
    .predict(

        sample_scaled,

        verbose=0

    )

)


sample_nn_index = np.argmax(

    sample_nn_probability,

    axis=1

)


sample_nn_label = (

    label_encoder
    .inverse_transform(
        sample_nn_index
    )[0]

)


print(
    "PSO H5 sample prediction:",
    sample_nn_label
)


# ======================================================================
# STEP 56: FINAL PSO SUMMARY
# ======================================================================

print(
    "\n"
    +
    "=" * 80
)

print(
    "PSO PROJECT COMPLETED SUCCESSFULLY"
)

print(
    "=" * 80
)


print(
    "\nBest PSO Fitness:"
)

print(
    round(
        global_best_score,
        4
    )
)


print(
    "\nSelected Features:"
)


for feature in selected_features:

    print(
        "-",
        feature
    )


print(
    "\nPSO Random Forest Accuracy:"
)

print(
    f"{pso_rf_accuracy * 100:.2f}%"
)


print(
    "\nPSO Neural Network Accuracy:"
)

print(
    f"{pso_nn_accuracy * 100:.2f}%"
)


print(
    "\nGenerated PSO Files:"
)


generated_files = [

    H5_PATH,

    PKL_PATH,

    YAML_PATH,

    JSON_PATH,

    ACCURACY_GRAPH_PATH,

    COMPARISON_GRAPH_PATH,

    HEATMAP_PATH,

    RESULT_CSV_PATH,

    RESULT_GRAPH_PATH,

    PREDICTION_CSV_PATH,

    PREDICTION_GRAPH_PATH,

    FEATURE_SELECTION_GRAPH_PATH,

    FEATURE_IMPORTANCE_GRAPH_PATH

]


for number, file_path in enumerate(

    generated_files,

    start=1

):

    print(
        f"{number}. {file_path}"
    )


print(
    "\nAll PSO outputs generated successfully."
)
```
