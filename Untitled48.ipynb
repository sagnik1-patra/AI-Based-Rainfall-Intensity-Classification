```python
import os
import json
import yaml
import joblib
import warnings
import random

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")

from sklearn.model_selection import (
    train_test_split
)

from sklearn.preprocessing import (
    StandardScaler,
    LabelEncoder
)

from sklearn.ensemble import (
    RandomForestClassifier
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

import tensorflow as tf

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Dense,
    Dropout
)

from tensorflow.keras.callbacks import (
    EarlyStopping
)


# ======================================================================
# STEP 2: RANDOM SEEDS
# ======================================================================

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
random.seed(RANDOM_STATE)
tf.random.set_seed(RANDOM_STATE)


# ======================================================================
# STEP 3: PROJECT PATHS
# ======================================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Rainfall Intensity Classification"
)

DATASET_PATH = os.path.join(
    BASE_DIR,
    "rainfall_0.csv"
)


# ----------------------------------------------------------------------
# MODEL FILES
# ----------------------------------------------------------------------

H5_PATH = os.path.join(
    BASE_DIR,
    "ais_rainfall_model.h5"
)

PKL_PATH = os.path.join(
    BASE_DIR,
    "ais_rainfall_model.pkl"
)

YAML_PATH = os.path.join(
    BASE_DIR,
    "ais_rainfall_config.yaml"
)

JSON_PATH = os.path.join(
    BASE_DIR,
    "ais_rainfall_metadata.json"
)


# ----------------------------------------------------------------------
# GRAPH FILES
# ----------------------------------------------------------------------

ACCURACY_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "ais_accuracy_graph.png"
)

COMPARISON_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "ais_comparison_graph.png"
)

HEATMAP_PATH = os.path.join(
    BASE_DIR,
    "ais_heatmap.png"
)

RESULT_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "ais_result_graph.png"
)

PREDICTION_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "ais_prediction_graph.png"
)

FEATURE_SELECTION_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "ais_feature_selection_graph.png"
)

FEATURE_IMPORTANCE_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "ais_feature_importance_graph.png"
)


# ----------------------------------------------------------------------
# CSV FILES
# ----------------------------------------------------------------------

RESULT_CSV_PATH = os.path.join(
    BASE_DIR,
    "ais_result.csv"
)

PREDICTION_CSV_PATH = os.path.join(
    BASE_DIR,
    "ais_prediction.csv"
)


os.makedirs(
    BASE_DIR,
    exist_ok=True
)


print("=" * 80)
print("AI-BASED RAINFALL INTENSITY CLASSIFICATION")
print("ARTIFICIAL IMMUNE SYSTEM (AIS) OPTIMIZATION")
print("=" * 80)


# ======================================================================
# STEP 4: LOAD DATASET
# ======================================================================

if not os.path.exists(DATASET_PATH):

    raise FileNotFoundError(
        f"\nDataset not found:\n{DATASET_PATH}"
    )


df = pd.read_csv(
    DATASET_PATH
)


print("\nDataset loaded successfully.")

print(
    "\nDataset shape:",
    df.shape
)


# ======================================================================
# STEP 5: CLEAN COLUMN NAMES
# ======================================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
)


print("\nColumns:")

for column in df.columns:

    print(column)


# ======================================================================
# STEP 6: IDENTIFY LOCATION COLUMNS
# ======================================================================

possible_location_columns = [

    "District",
    "Taluka",
    "Taluk",
    "Circle",
    "Station",
    "Location",
    "Village"

]


location_columns = [

    col

    for col in df.columns

    if col.lower() in [

        item.lower()

        for item in possible_location_columns

    ]

]


print(
    "\nLocation columns:",
    location_columns
)


# ======================================================================
# STEP 7: IDENTIFY RAINFALL COLUMNS
# ======================================================================

candidate_columns = [

    col

    for col in df.columns

    if col not in location_columns

]


rainfall_columns = []


for col in candidate_columns:

    converted = pd.to_numeric(
        df[col],
        errors="coerce"
    )

    numeric_count = (
        converted
        .notna()
        .sum()
    )

    minimum_required = max(
        3,
        int(len(df) * 0.30)
    )

    if numeric_count >= minimum_required:

        rainfall_columns.append(
            col
        )


if len(rainfall_columns) == 0:

    raise ValueError(
        "No rainfall measurement columns detected."
    )


print(
    "\nNumber of rainfall columns:",
    len(rainfall_columns)
)


# ======================================================================
# STEP 8: CONVERT RAINFALL DATA TO NUMERIC
# ======================================================================

for col in rainfall_columns:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )


df[rainfall_columns] = (

    df[rainfall_columns]

    .replace(
        [np.inf, -np.inf],
        np.nan
    )

    .fillna(0)

)


df[rainfall_columns] = (

    df[rainfall_columns]
    .clip(lower=0)

)


print(
    "\nRainfall data cleaned successfully."
)


# ======================================================================
# STEP 9: FEATURE ENGINEERING
# ======================================================================

rainfall_data = (
    df[rainfall_columns]
)


df["Total_Rainfall"] = (
    rainfall_data.sum(axis=1)
)


df["Average_Rainfall"] = (
    rainfall_data.mean(axis=1)
)


df["Maximum_Rainfall"] = (
    rainfall_data.max(axis=1)
)


df["Minimum_Rainfall"] = (
    rainfall_data.min(axis=1)
)


df["Rainfall_Std"] = (
    rainfall_data.std(axis=1)
)


df["Rainy_Days"] = (
    (rainfall_data > 0)
    .sum(axis=1)
)


df["Moderate_Rain_Days"] = (
    (rainfall_data >= 15.6)
    .sum(axis=1)
)


df["Heavy_Rain_Days"] = (
    (rainfall_data >= 64.5)
    .sum(axis=1)
)


df["Very_Heavy_Rain_Days"] = (
    (rainfall_data >= 115.6)
    .sum(axis=1)
)


df["Extreme_Rain_Days"] = (
    (rainfall_data >= 204.5)
    .sum(axis=1)
)


df["Rainfall_Range"] = (

    df["Maximum_Rainfall"]

    -

    df["Minimum_Rainfall"]

)


df["Rainfall_CV"] = (

    df["Rainfall_Std"]

    /

    (
        df["Average_Rainfall"]
        +
        1e-6
    )

)


# ======================================================================
# STEP 10: MONTH-WISE FEATURES
# ======================================================================

august_columns = [

    col

    for col in rainfall_columns

    if "aug" in col.lower()

]


september_columns = [

    col

    for col in rainfall_columns

    if "sep" in col.lower()

]


if len(august_columns) > 0:

    df["August_Total_Rainfall"] = (

        df[august_columns]
        .sum(axis=1)

    )

    df["August_Average_Rainfall"] = (

        df[august_columns]
        .mean(axis=1)

    )

else:

    df["August_Total_Rainfall"] = 0

    df["August_Average_Rainfall"] = 0


if len(september_columns) > 0:

    df["September_Total_Rainfall"] = (

        df[september_columns]
        .sum(axis=1)

    )

    df["September_Average_Rainfall"] = (

        df[september_columns]
        .mean(axis=1)

    )

else:

    df["September_Total_Rainfall"] = 0

    df["September_Average_Rainfall"] = 0


print(
    "\nFeature engineering completed."
)


# ======================================================================
# STEP 11: CREATE TARGET CLASSES
# ======================================================================

lower_threshold = (
    df["Total_Rainfall"]
    .quantile(0.33)
)

upper_threshold = (
    df["Total_Rainfall"]
    .quantile(0.66)
)


def classify_rainfall(value):

    if value <= lower_threshold:

        return "Low"

    elif value <= upper_threshold:

        return "Moderate"

    else:

        return "High"


df["Rainfall_Intensity"] = (

    df["Total_Rainfall"]
    .apply(
        classify_rainfall
    )

)


print(
    "\nRainfall thresholds:"
)

print(
    "Low <=",
    round(lower_threshold, 2)
)

print(
    "Moderate <=",
    round(upper_threshold, 2)
)

print(
    "High >",
    round(upper_threshold, 2)
)


print(
    "\nClass distribution:"
)

print(
    df["Rainfall_Intensity"]
    .value_counts()
)


# ======================================================================
# STEP 12: DEFINE ALL CANDIDATE FEATURES
# ======================================================================

all_features = [

    "Average_Rainfall",

    "Maximum_Rainfall",

    "Minimum_Rainfall",

    "Rainfall_Std",

    "Rainy_Days",

    "Moderate_Rain_Days",

    "Heavy_Rain_Days",

    "Very_Heavy_Rain_Days",

    "Extreme_Rain_Days",

    "Rainfall_Range",

    "Rainfall_CV",

    "August_Total_Rainfall",

    "August_Average_Rainfall",

    "September_Total_Rainfall",

    "September_Average_Rainfall"

]


X = (
    df[all_features]
    .copy()
)


X = (

    X.replace(
        [np.inf, -np.inf],
        np.nan
    )

    .fillna(0)

)


y = (
    df["Rainfall_Intensity"]
    .copy()
)


print(
    "\nTotal candidate features:",
    len(all_features)
)


# ======================================================================
# STEP 13: LABEL ENCODING
# ======================================================================

label_encoder = (
    LabelEncoder()
)


y_encoded = (
    label_encoder
    .fit_transform(y)
)


print(
    "\nClasses:"
)


for i, label in enumerate(
    label_encoder.classes_
):

    print(
        i,
        "=",
        label
    )


# ======================================================================
# STEP 14: CREATE TRAIN AND TEST DATA
# ======================================================================

indices = np.arange(
    len(df)
)


(
    X_train_full,
    X_test_full,
    y_train,
    y_test,
    train_indices,
    test_indices

) = train_test_split(

    X,

    y_encoded,

    indices,

    test_size=0.25,

    random_state=RANDOM_STATE,

    stratify=y_encoded

)


print(
    "\nTraining samples:",
    len(X_train_full)
)

print(
    "Testing samples:",
    len(X_test_full)
)


# ======================================================================
# STEP 15: CREATE AIS TRAINING AND VALIDATION SET
# ======================================================================

try:

    (
        X_ais_train,
        X_ais_validation,
        y_ais_train,
        y_ais_validation

    ) = train_test_split(

        X_train_full,

        y_train,

        test_size=0.25,

        random_state=RANDOM_STATE,

        stratify=y_train

    )

except ValueError:

    (
        X_ais_train,
        X_ais_validation,
        y_ais_train,
        y_ais_validation

    ) = train_test_split(

        X_train_full,

        y_train,

        test_size=0.25,

        random_state=RANDOM_STATE

    )


print(
    "\nAIS training samples:",
    len(X_ais_train)
)

print(
    "AIS validation samples:",
    len(X_ais_validation)
)


# ======================================================================
# STEP 16: AIS CONFIGURATION
# ======================================================================

NUMBER_OF_FEATURES = len(
    all_features
)

POPULATION_SIZE = 20

GENERATIONS = 20

ELITE_SIZE = 5

CLONES_PER_ELITE = 4

MUTATION_RATE = 0.15

RANDOM_NEW_ANTIBODIES = 5


print(
    "\nAIS configuration:"
)

print(
    "Population size:",
    POPULATION_SIZE
)

print(
    "Generations:",
    GENERATIONS
)

print(
    "Elite antibodies:",
    ELITE_SIZE
)

print(
    "Clones per elite:",
    CLONES_PER_ELITE
)

print(
    "Mutation rate:",
    MUTATION_RATE
)


# ======================================================================
# STEP 17: CREATE RANDOM ANTIBODY
# ======================================================================

def create_antibody():

    antibody = np.random.randint(

        0,

        2,

        size=NUMBER_OF_FEATURES

    )

    # Make sure at least one feature is selected
    if np.sum(antibody) == 0:

        random_index = np.random.randint(
            0,
            NUMBER_OF_FEATURES
        )

        antibody[random_index] = 1

    return antibody


# ======================================================================
# STEP 18: AIS FITNESS FUNCTION
# ======================================================================

def evaluate_antibody(antibody):

    selected_indices = np.where(
        antibody == 1
    )[0]

    if len(selected_indices) == 0:

        return 0.0

    selected_features = [

        all_features[i]

        for i in selected_indices

    ]

    X_train_selected = (

        X_ais_train[
            selected_features
        ]

    )

    X_validation_selected = (

        X_ais_validation[
            selected_features
        ]

    )

    model = RandomForestClassifier(

        n_estimators=120,

        random_state=RANDOM_STATE,

        class_weight="balanced",

        n_jobs=-1

    )

    model.fit(

        X_train_selected,

        y_ais_train

    )

    prediction = model.predict(

        X_validation_selected

    )

    accuracy = accuracy_score(

        y_ais_validation,

        prediction

    )

    # --------------------------------------------------------------
    # Small feature penalty
    #
    # This encourages AIS to find compact feature subsets rather than
    # simply selecting every feature.
    # --------------------------------------------------------------

    feature_ratio = (

        len(selected_features)

        /

        NUMBER_OF_FEATURES

    )

    fitness = (

        accuracy

        -

        (
            0.01
            *
            feature_ratio
        )

    )

    return fitness


# ======================================================================
# STEP 19: MUTATION FUNCTION
# ======================================================================

def mutate_antibody(

    antibody,

    mutation_rate=MUTATION_RATE

):

    mutated = antibody.copy()

    for i in range(
        len(mutated)
    ):

        if np.random.random() < mutation_rate:

            mutated[i] = (
                1
                -
                mutated[i]
            )

    # Ensure at least one selected feature
    if np.sum(mutated) == 0:

        random_index = np.random.randint(

            0,

            len(mutated)

        )

        mutated[random_index] = 1

    return mutated


# ======================================================================
# STEP 20: INITIALIZE AIS POPULATION
# ======================================================================

population = [

    create_antibody()

    for _ in range(
        POPULATION_SIZE
    )

]


ais_history = []

best_antibody = None

best_fitness = -np.inf


print(
    "\n"
    +
    "=" * 80
)

print(
    "STARTING AIS FEATURE SELECTION"
)

print(
    "=" * 80
)


# ======================================================================
# STEP 21: RUN AIS / CLONAL SELECTION
# ======================================================================

for generation in range(
    GENERATIONS
):

    # --------------------------------------------------------------
    # Evaluate current population
    # --------------------------------------------------------------

    evaluated_population = []

    for antibody in population:

        fitness = evaluate_antibody(
            antibody
        )

        evaluated_population.append(

            (
                fitness,
                antibody.copy()
            )

        )


    # --------------------------------------------------------------
    # Sort by fitness
    # --------------------------------------------------------------

    evaluated_population.sort(

        key=lambda item: item[0],

        reverse=True

    )


    generation_best_fitness = (

        evaluated_population[0][0]

    )

    generation_best_antibody = (

        evaluated_population[0][1]
        .copy()

    )


    # --------------------------------------------------------------
    # Update global best
    # --------------------------------------------------------------

    if generation_best_fitness > best_fitness:

        best_fitness = (
            generation_best_fitness
        )

        best_antibody = (
            generation_best_antibody
            .copy()
        )


    ais_history.append(
        best_fitness
    )


    selected_count = int(
        np.sum(
            generation_best_antibody
        )
    )


    print(
        f"Generation {generation + 1:02d}/{GENERATIONS}"
        f" | Best Fitness: {best_fitness:.4f}"
        f" | Selected Features: {selected_count}"
    )


    # --------------------------------------------------------------
    # Select elite antibodies
    # --------------------------------------------------------------

    elites = [

        item[1].copy()

        for item in
        evaluated_population[
            :ELITE_SIZE
        ]

    ]


    # --------------------------------------------------------------
    # Clonal expansion
    # --------------------------------------------------------------

    clones = []

    for rank, elite in enumerate(
        elites
    ):

        # Better antibodies receive slightly lower mutation
        adaptive_mutation_rate = (

            MUTATION_RATE

            *
            (
                (rank + 1)
                /
                ELITE_SIZE
            )

        )

        for _ in range(
            CLONES_PER_ELITE
        ):

            clone = mutate_antibody(

                elite,

                mutation_rate=
                adaptive_mutation_rate

            )

            clones.append(
                clone
            )


    # --------------------------------------------------------------
    # Evaluate clones
    # --------------------------------------------------------------

    evaluated_clones = []

    for clone in clones:

        clone_fitness = evaluate_antibody(
            clone
        )

        evaluated_clones.append(

            (
                clone_fitness,
                clone.copy()
            )

        )


    # --------------------------------------------------------------
    # Combine parents and clones
    # --------------------------------------------------------------

    combined_population = (

        evaluated_population

        +

        evaluated_clones

    )


    combined_population.sort(

        key=lambda item: item[0],

        reverse=True

    )


    # --------------------------------------------------------------
    # Retain strongest antibodies
    # --------------------------------------------------------------

    survivor_count = (

        POPULATION_SIZE

        -

        RANDOM_NEW_ANTIBODIES

    )


    population = [

        item[1].copy()

        for item in
        combined_population[
            :survivor_count
        ]

    ]


    # --------------------------------------------------------------
    # Add random antibodies for diversity
    # --------------------------------------------------------------

    while len(population) < POPULATION_SIZE:

        population.append(
            create_antibody()
        )


# ======================================================================
# STEP 22: EXTRACT BEST AIS FEATURES
# ======================================================================

selected_feature_indices = np.where(

    best_antibody == 1

)[0]


selected_features = [

    all_features[i]

    for i in selected_feature_indices

]


print(
    "\n"
    +
    "=" * 80
)

print(
    "AIS FEATURE SELECTION COMPLETED"
)

print(
    "=" * 80
)


print(
    "\nBest AIS fitness:",
    round(
        best_fitness,
        4
    )
)


print(
    "\nNumber of selected features:",
    len(selected_features)
)


print(
    "\nSelected features:"
)


for number, feature in enumerate(

    selected_features,

    start=1

):

    print(
        f"{number}. {feature}"
    )


# ======================================================================
# STEP 23: AIS FEATURE SELECTION GRAPH
# ======================================================================

plt.figure(
    figsize=(10, 6)
)


plt.plot(

    range(
        1,
        len(ais_history) + 1
    ),

    ais_history,

    marker="o",

    linewidth=2

)


plt.title(

    "AIS Feature Selection Optimization",

    fontsize=14,

    fontweight="bold"

)


plt.xlabel(
    "Generation"
)


plt.ylabel(
    "Best Fitness"
)


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    FEATURE_SELECTION_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 24: PREPARE AIS SELECTED DATA
# ======================================================================

X_selected = (

    X[
        selected_features
    ]
    .copy()

)


X_train = (

    X_train_full[
        selected_features
    ]
    .copy()

)


X_test = (

    X_test_full[
        selected_features
    ]
    .copy()

)


# ======================================================================
# STEP 25: FEATURE SCALING
# ======================================================================

scaler = (
    StandardScaler()
)


X_train_scaled = scaler.fit_transform(
    X_train
)


X_test_scaled = scaler.transform(
    X_test
)


# ======================================================================
# STEP 26: TRAIN AIS RANDOM FOREST
# ======================================================================

print(
    "\nTraining AIS-optimized Random Forest..."
)


ais_rf_model = RandomForestClassifier(

    n_estimators=500,

    class_weight="balanced",

    random_state=RANDOM_STATE,

    n_jobs=-1

)


ais_rf_model.fit(

    X_train,

    y_train

)


ais_rf_prediction = ais_rf_model.predict(

    X_test

)


# ======================================================================
# STEP 27: AIS RANDOM FOREST METRICS
# ======================================================================

ais_rf_accuracy = accuracy_score(

    y_test,

    ais_rf_prediction

)


ais_rf_precision = precision_score(

    y_test,

    ais_rf_prediction,

    average="weighted",

    zero_division=0

)


ais_rf_recall = recall_score(

    y_test,

    ais_rf_prediction,

    average="weighted",

    zero_division=0

)


ais_rf_f1 = f1_score(

    y_test,

    ais_rf_prediction,

    average="weighted",

    zero_division=0

)


print(
    "\nAIS Random Forest Accuracy:",
    f"{ais_rf_accuracy * 100:.2f}%"
)


# ======================================================================
# STEP 28: SAVE AIS RANDOM FOREST MODEL
# ======================================================================

ais_model_package = {

    "model":
        ais_rf_model,

    "scaler":
        scaler,

    "label_encoder":
        label_encoder,

    "selected_features":
        selected_features,

    "all_features":
        all_features,

    "best_antibody":
        best_antibody,

    "best_fitness":
        best_fitness,

    "rainfall_columns":
        rainfall_columns,

    "lower_threshold":
        lower_threshold,

    "upper_threshold":
        upper_threshold

}


joblib.dump(

    ais_model_package,

    PKL_PATH

)


print(
    "\nAIS PKL saved:"
)

print(
    PKL_PATH
)


# ======================================================================
# STEP 29: PREPARE NEURAL NETWORK LABELS
# ======================================================================

number_of_classes = len(
    label_encoder.classes_
)


y_train_nn = (

    tf.keras.utils
    .to_categorical(

        y_train,

        num_classes=
        number_of_classes

    )

)


y_test_nn = (

    tf.keras.utils
    .to_categorical(

        y_test,

        num_classes=
        number_of_classes

    )

)


# ======================================================================
# STEP 30: BUILD AIS NEURAL NETWORK
# ======================================================================

print(
    "\nTraining AIS-optimized Neural Network..."
)


ais_nn_model = Sequential([

    tf.keras.layers.Input(

        shape=(
            len(selected_features),
        )

    ),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(
        0.20
    ),

    Dense(
        32,
        activation="relu"
    ),

    Dropout(
        0.15
    ),

    Dense(
        16,
        activation="relu"
    ),

    Dense(

        number_of_classes,

        activation="softmax"

    )

])


ais_nn_model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="categorical_crossentropy",

    metrics=[
        "accuracy"
    ]

)


# ======================================================================
# STEP 31: EARLY STOPPING
# ======================================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=20,

    restore_best_weights=True,

    verbose=1

)


# ======================================================================
# STEP 32: TRAIN AIS NEURAL NETWORK
# ======================================================================

history = ais_nn_model.fit(

    X_train_scaled,

    y_train_nn,

    validation_split=0.20,

    epochs=200,

    batch_size=8,

    callbacks=[
        early_stopping
    ],

    verbose=1

)


# ======================================================================
# STEP 33: AIS NEURAL NETWORK PREDICTION
# ======================================================================

ais_nn_probability = ais_nn_model.predict(

    X_test_scaled,

    verbose=0

)


ais_nn_prediction = np.argmax(

    ais_nn_probability,

    axis=1

)


# ======================================================================
# STEP 34: AIS NEURAL NETWORK METRICS
# ======================================================================

ais_nn_accuracy = accuracy_score(

    y_test,

    ais_nn_prediction

)


ais_nn_precision = precision_score(

    y_test,

    ais_nn_prediction,

    average="weighted",

    zero_division=0

)


ais_nn_recall = recall_score(

    y_test,

    ais_nn_prediction,

    average="weighted",

    zero_division=0

)


ais_nn_f1 = f1_score(

    y_test,

    ais_nn_prediction,

    average="weighted",

    zero_division=0

)


print(
    "\nAIS Neural Network Accuracy:",
    f"{ais_nn_accuracy * 100:.2f}%"
)


# ======================================================================
# STEP 35: SAVE AIS H5 MODEL
# ======================================================================

ais_nn_model.save(
    H5_PATH
)


print(
    "\nAIS H5 model saved:"
)

print(
    H5_PATH
)


# ======================================================================
# STEP 36: AIS ACCURACY GRAPH
# ======================================================================

epochs = range(

    1,

    len(
        history.history["accuracy"]
    )
    +
    1

)


plt.figure(
    figsize=(10, 6)
)


plt.plot(

    epochs,

    history.history[
        "accuracy"
    ],

    marker="o",

    markersize=3,

    label="Training Accuracy"

)


plt.plot(

    epochs,

    history.history[
        "val_accuracy"
    ],

    marker="s",

    markersize=3,

    label="Validation Accuracy"

)


plt.title(

    "AIS Neural Network Training and Validation Accuracy",

    fontsize=14,

    fontweight="bold"

)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "Accuracy"
)


plt.ylim(
    0,
    1.05
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    ACCURACY_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 37: MODEL COMPARISON DATA
# ======================================================================

comparison_data = pd.DataFrame({

    "Model": [

        "AIS Random Forest",

        "AIS Neural Network"

    ],

    "Accuracy": [

        ais_rf_accuracy,

        ais_nn_accuracy

    ],

    "Precision": [

        ais_rf_precision,

        ais_nn_precision

    ],

    "Recall": [

        ais_rf_recall,

        ais_nn_recall

    ],

    "F1 Score": [

        ais_rf_f1,

        ais_nn_f1

    ]

})


# ======================================================================
# STEP 38: AIS COMPARISON GRAPH
# ======================================================================

comparison_plot = (

    comparison_data

    .set_index(
        "Model"
    )

    *

    100

)


ax = comparison_plot.plot(

    kind="bar",

    figsize=(11, 6),

    width=0.75

)


plt.title(

    "AIS-Optimized Model Performance Comparison",

    fontsize=14,

    fontweight="bold"

)


plt.xlabel(
    "Model"
)


plt.ylabel(
    "Score (%)"
)


plt.ylim(
    0,
    105
)


plt.xticks(
    rotation=0
)


plt.grid(

    axis="y",

    alpha=0.3

)


plt.legend(
    title="Metrics"
)


for container in ax.containers:

    ax.bar_label(

        container,

        fmt="%.1f",

        padding=3,

        fontsize=8

    )


plt.tight_layout()


plt.savefig(

    COMPARISON_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 39: AIS CONFUSION MATRIX
# ======================================================================

cm = confusion_matrix(

    y_test,

    ais_rf_prediction

)


fig, ax = plt.subplots(

    figsize=(8, 6)

)


image = ax.imshow(

    cm,

    interpolation="nearest",

    cmap="Blues"

)


plt.colorbar(
    image
)


classes = (
    label_encoder.classes_
)


ax.set(

    xticks=np.arange(
        len(classes)
    ),

    yticks=np.arange(
        len(classes)
    ),

    xticklabels=classes,

    yticklabels=classes,

    xlabel="Predicted Rainfall Intensity",

    ylabel="Actual Rainfall Intensity",

    title="AIS Random Forest Confusion Matrix"

)


threshold = (

    cm.max() / 2

    if cm.size > 0

    else 0

)


for i in range(
    cm.shape[0]
):

    for j in range(
        cm.shape[1]
    ):

        ax.text(

            j,

            i,

            str(
                cm[i, j]
            ),

            ha="center",

            va="center",

            color=(

                "white"

                if cm[i, j] > threshold

                else "black"

            ),

            fontsize=12,

            fontweight="bold"

        )


plt.tight_layout()


plt.savefig(

    HEATMAP_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 40: FEATURE IMPORTANCE
# ======================================================================

feature_importance = pd.DataFrame({

    "Feature":
        selected_features,

    "Importance":
        ais_rf_model.feature_importances_

})


feature_importance = (

    feature_importance

    .sort_values(

        "Importance",

        ascending=True

    )

)


plt.figure(
    figsize=(10, 7)
)


plt.barh(

    feature_importance[
        "Feature"
    ],

    feature_importance[
        "Importance"
    ]

)


plt.title(

    "AIS Selected Feature Importance",

    fontsize=14,

    fontweight="bold"

)


plt.xlabel(
    "Importance Score"
)


plt.ylabel(
    "Selected Feature"
)


plt.grid(

    axis="x",

    alpha=0.3

)


plt.tight_layout()


plt.savefig(

    FEATURE_IMPORTANCE_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 41: CREATE AIS RESULT CSV
# ======================================================================

actual_labels = (

    label_encoder
    .inverse_transform(
        y_test
    )

)


rf_labels = (

    label_encoder
    .inverse_transform(
        ais_rf_prediction
    )

)


nn_labels = (

    label_encoder
    .inverse_transform(
        ais_nn_prediction
    )

)


test_df = (

    df.iloc[
        test_indices
    ]

    .copy()

    .reset_index(
        drop=True
    )

)


result_output = pd.DataFrame()


for col in location_columns:

    result_output[col] = (
        test_df[col]
    )


result_output[
    "Total_Rainfall"
] = test_df[
    "Total_Rainfall"
]


result_output[
    "Average_Rainfall"
] = test_df[
    "Average_Rainfall"
]


result_output[
    "Maximum_Rainfall"
] = test_df[
    "Maximum_Rainfall"
]


result_output[
    "Rainy_Days"
] = test_df[
    "Rainy_Days"
]


result_output[
    "Heavy_Rain_Days"
] = test_df[
    "Heavy_Rain_Days"
]


result_output[
    "Actual_Intensity"
] = actual_labels


result_output[
    "AIS_RF_Prediction"
] = rf_labels


result_output[
    "AIS_NN_Prediction"
] = nn_labels


result_output[
    "AIS_RF_Correct"
] = (

    result_output[
        "Actual_Intensity"
    ]

    ==

    result_output[
        "AIS_RF_Prediction"
    ]

)


result_output[
    "AIS_NN_Correct"
] = (

    result_output[
        "Actual_Intensity"
    ]

    ==

    result_output[
        "AIS_NN_Prediction"
    ]

)


result_output.to_csv(

    RESULT_CSV_PATH,

    index=False

)


print(
    "\nAIS result CSV saved:"
)

print(
    RESULT_CSV_PATH
)


print(
    "\nAIS Result Preview:"
)

print(
    result_output.head(10)
)


# ======================================================================
# STEP 42: AIS RESULT GRAPH
# ======================================================================

actual_counts = (

    result_output[
        "Actual_Intensity"
    ]

    .value_counts()

    .reindex(
        classes,
        fill_value=0
    )

)


rf_counts = (

    result_output[
        "AIS_RF_Prediction"
    ]

    .value_counts()

    .reindex(
        classes,
        fill_value=0
    )

)


nn_counts = (

    result_output[
        "AIS_NN_Prediction"
    ]

    .value_counts()

    .reindex(
        classes,
        fill_value=0
    )

)


x_position = np.arange(
    len(classes)
)


width = 0.25


plt.figure(
    figsize=(10, 6)
)


plt.bar(

    x_position - width,

    actual_counts.values,

    width,

    label="Actual"

)


plt.bar(

    x_position,

    rf_counts.values,

    width,

    label="AIS Random Forest"

)


plt.bar(

    x_position + width,

    nn_counts.values,

    width,

    label="AIS Neural Network"

)


plt.title(

    "AIS Actual vs Predicted Rainfall Intensity",

    fontsize=14,

    fontweight="bold"

)


plt.xlabel(
    "Rainfall Intensity"
)


plt.ylabel(
    "Number of Records"
)


plt.xticks(

    x_position,

    classes

)


plt.legend()


plt.grid(

    axis="y",

    alpha=0.3

)


plt.tight_layout()


plt.savefig(

    RESULT_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 43: PREDICT COMPLETE DATASET
# ======================================================================

X_all_selected = (

    X[
        selected_features
    ]
    .copy()

)


# Random Forest
ais_rf_all_prediction = (

    ais_rf_model
    .predict(
        X_all_selected
    )

)


ais_rf_all_probability = (

    ais_rf_model
    .predict_proba(
        X_all_selected
    )

)


ais_rf_all_labels = (

    label_encoder
    .inverse_transform(
        ais_rf_all_prediction
    )

)


# Neural Network
X_all_scaled = (

    scaler
    .transform(
        X_all_selected
    )

)


ais_nn_all_probability = (

    ais_nn_model
    .predict(

        X_all_scaled,

        verbose=0

    )

)


ais_nn_all_prediction = np.argmax(

    ais_nn_all_probability,

    axis=1

)


ais_nn_all_labels = (

    label_encoder
    .inverse_transform(
        ais_nn_all_prediction
    )

)


# ======================================================================
# STEP 44: CREATE AIS PREDICTION CSV
# ======================================================================

prediction_output = pd.DataFrame()


for col in location_columns:

    prediction_output[col] = (
        df[col]
    )


prediction_output[
    "Total_Rainfall"
] = df[
    "Total_Rainfall"
]


prediction_output[
    "Average_Rainfall"
] = df[
    "Average_Rainfall"
]


prediction_output[
    "Maximum_Rainfall"
] = df[
    "Maximum_Rainfall"
]


prediction_output[
    "Rainfall_Std"
] = df[
    "Rainfall_Std"
]


prediction_output[
    "Rainy_Days"
] = df[
    "Rainy_Days"
]


prediction_output[
    "Heavy_Rain_Days"
] = df[
    "Heavy_Rain_Days"
]


prediction_output[
    "Actual_Intensity"
] = df[
    "Rainfall_Intensity"
]


prediction_output[
    "AIS_RF_Prediction"
] = ais_rf_all_labels


prediction_output[
    "AIS_NN_Prediction"
] = ais_nn_all_labels


prediction_output[
    "AIS_RF_Confidence"
] = (

    np.max(

        ais_rf_all_probability,

        axis=1

    )

    *

    100

)


prediction_output[
    "AIS_NN_Confidence"
] = (

    np.max(

        ais_nn_all_probability,

        axis=1

    )

    *

    100

)


prediction_output.to_csv(

    PREDICTION_CSV_PATH,

    index=False

)


print(
    "\nAIS prediction CSV saved:"
)

print(
    PREDICTION_CSV_PATH
)


print(
    "\nPrediction Preview:"
)

print(
    prediction_output.head(10)
)


# ======================================================================
# STEP 45: AIS PREDICTION GRAPH
# ======================================================================

prediction_counts = (

    prediction_output[
        "AIS_RF_Prediction"
    ]

    .value_counts()

    .reindex(
        classes,
        fill_value=0
    )

)


plt.figure(
    figsize=(9, 6)
)


bars = plt.bar(

    prediction_counts.index,

    prediction_counts.values

)


plt.title(

    "AIS Predicted Rainfall Intensity Distribution",

    fontsize=14,

    fontweight="bold"

)


plt.xlabel(
    "Rainfall Intensity"
)


plt.ylabel(
    "Number of Locations"
)


plt.grid(

    axis="y",

    alpha=0.3

)


for bar in bars:

    height = bar.get_height()

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        height,

        str(
            int(height)
        ),

        ha="center",

        va="bottom",

        fontweight="bold"

    )


plt.tight_layout()


plt.savefig(

    PREDICTION_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 46: SAVE YAML CONFIGURATION
# ======================================================================

yaml_config = {

    "project": {

        "name":
            "AI-Based Rainfall Intensity Classification",

        "optimization":
            "Artificial Immune System (AIS)",

        "ais_method":
            "Clonal Selection Feature Optimization"

    },

    "dataset": {

        "filename":
            os.path.basename(
                DATASET_PATH
            ),

        "rows":
            int(
                df.shape[0]
            ),

        "rainfall_columns":
            int(
                len(
                    rainfall_columns
                )
            )

    },

    "ais": {

        "population_size":
            POPULATION_SIZE,

        "generations":
            GENERATIONS,

        "elite_size":
            ELITE_SIZE,

        "clones_per_elite":
            CLONES_PER_ELITE,

        "mutation_rate":
            MUTATION_RATE,

        "best_fitness":
            float(
                best_fitness
            ),

        "selected_feature_count":
            len(
                selected_features
            ),

        "selected_features":
            selected_features

    },

    "target": {

        "column":
            "Rainfall_Intensity",

        "classes":
            label_encoder
            .classes_
            .tolist()

    },

    "models": {

        "ais_random_forest": {

            "accuracy":
                float(
                    ais_rf_accuracy
                ),

            "precision":
                float(
                    ais_rf_precision
                ),

            "recall":
                float(
                    ais_rf_recall
                ),

            "f1_score":
                float(
                    ais_rf_f1
                )

        },

        "ais_neural_network": {

            "accuracy":
                float(
                    ais_nn_accuracy
                ),

            "precision":
                float(
                    ais_nn_precision
                ),

            "recall":
                float(
                    ais_nn_recall
                ),

            "f1_score":
                float(
                    ais_nn_f1
                )

        }

    }

}


with open(

    YAML_PATH,

    "w",

    encoding="utf-8"

) as file:

    yaml.safe_dump(

        yaml_config,

        file,

        sort_keys=False,

        allow_unicode=True

    )


print(
    "\nAIS YAML saved:"
)

print(
    YAML_PATH
)


# ======================================================================
# STEP 47: SAVE JSON METADATA
# ======================================================================

json_metadata = {

    "project_name":
        "AI-Based Rainfall Intensity Classification",

    "optimization_algorithm":
        "Artificial Immune System",

    "ais_implementation":
        "Clonal Selection Feature Optimization",

    "dataset":
        os.path.basename(
            DATASET_PATH
        ),

    "dataset_rows":
        int(
            len(df)
        ),

    "candidate_features":
        all_features,

    "selected_features":
        selected_features,

    "selected_feature_count":
        int(
            len(
                selected_features
            )
        ),

    "best_antibody":
        best_antibody
        .astype(int)
        .tolist(),

    "best_fitness":
        float(
            best_fitness
        ),

    "ais_parameters": {

        "population_size":
            POPULATION_SIZE,

        "generations":
            GENERATIONS,

        "elite_size":
            ELITE_SIZE,

        "clones_per_elite":
            CLONES_PER_ELITE,

        "mutation_rate":
            MUTATION_RATE

    },

    "rainfall_thresholds": {

        "low_upper":
            float(
                lower_threshold
            ),

        "moderate_upper":
            float(
                upper_threshold
            )

    },

    "ais_random_forest": {

        "accuracy":
            float(
                ais_rf_accuracy
            ),

        "precision":
            float(
                ais_rf_precision
            ),

        "recall":
            float(
                ais_rf_recall
            ),

        "f1_score":
            float(
                ais_rf_f1
            )

    },

    "ais_neural_network": {

        "accuracy":
            float(
                ais_nn_accuracy
            ),

        "precision":
            float(
                ais_nn_precision
            ),

        "recall":
            float(
                ais_nn_recall
            ),

        "f1_score":
            float(
                ais_nn_f1
            )

    }

}


with open(

    JSON_PATH,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        json_metadata,

        file,

        indent=4

    )


print(
    "\nAIS JSON saved:"
)

print(
    JSON_PATH
)


# ======================================================================
# STEP 48: RANDOM FOREST CLASSIFICATION REPORT
# ======================================================================

print(
    "\n"
    +
    "=" * 80
)

print(
    "AIS RANDOM FOREST CLASSIFICATION REPORT"
)

print(
    "=" * 80
)


print(

    classification_report(

        y_test,

        ais_rf_prediction,

        target_names=
        label_encoder.classes_,

        zero_division=0

    )

)


# ======================================================================
# STEP 49: NEURAL NETWORK CLASSIFICATION REPORT
# ======================================================================

print(
    "\n"
    +
    "=" * 80
)

print(
    "AIS NEURAL NETWORK CLASSIFICATION REPORT"
)

print(
    "=" * 80
)


print(

    classification_report(

        y_test,

        ais_nn_prediction,

        target_names=
        label_encoder.classes_,

        zero_division=0

    )

)


# ======================================================================
# STEP 50: TEST SAVED PKL
# ======================================================================

print(
    "\nTesting saved AIS PKL model..."
)


loaded_package = joblib.load(
    PKL_PATH
)


loaded_rf_model = (
    loaded_package[
        "model"
    ]
)


sample = (

    X_selected
    .iloc[
        [0]
    ]

)


sample_prediction = (
    loaded_rf_model
    .predict(
        sample
    )
)


sample_label = (

    label_encoder
    .inverse_transform(
        sample_prediction
    )[0]

)


print(
    "AIS PKL sample prediction:",
    sample_label
)


# ======================================================================
# STEP 51: TEST SAVED H5
# ======================================================================

print(
    "\nTesting saved AIS H5 model..."
)


loaded_nn_model = (

    tf.keras.models
    .load_model(
        H5_PATH
    )

)


sample_scaled = (
    scaler.transform(
        sample
    )
)


sample_nn_probability = (

    loaded_nn_model
    .predict(

        sample_scaled,

        verbose=0

    )

)


sample_nn_index = np.argmax(

    sample_nn_probability,

    axis=1

)


sample_nn_label = (

    label_encoder
    .inverse_transform(
        sample_nn_index
    )[0]

)


print(
    "AIS H5 sample prediction:",
    sample_nn_label
)


# ======================================================================
# STEP 52: FINAL SUMMARY
# ======================================================================

print(
    "\n"
    +
    "=" * 80
)

print(
    "AIS PROJECT COMPLETED SUCCESSFULLY"
)

print(
    "=" * 80
)


print(
    "\nBest AIS Fitness:"
)

print(
    round(
        best_fitness,
        4
    )
)


print(
    "\nSelected Features:"
)


for feature in selected_features:

    print(
        "-",
        feature
    )


print(
    "\nAIS Random Forest Accuracy:"
)

print(
    f"{ais_rf_accuracy * 100:.2f}%"
)


print(
    "\nAIS Neural Network Accuracy:"
)

print(
    f"{ais_nn_accuracy * 100:.2f}%"
)


print(
    "\nGenerated AIS Files:"
)


generated_files = [

    H5_PATH,

    PKL_PATH,

    YAML_PATH,

    JSON_PATH,

    ACCURACY_GRAPH_PATH,

    COMPARISON_GRAPH_PATH,

    HEATMAP_PATH,

    RESULT_CSV_PATH,

    RESULT_GRAPH_PATH,

    PREDICTION_CSV_PATH,

    PREDICTION_GRAPH_PATH,

    FEATURE_SELECTION_GRAPH_PATH,

    FEATURE_IMPORTANCE_GRAPH_PATH

]


for number, file_path in enumerate(

    generated_files,

    start=1

):

    print(
        f"{number}. {file_path}"
    )


print(
    "\nAll AIS outputs were generated successfully."
)
```
