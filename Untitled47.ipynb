```python
import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping


# ======================================================================
# STEP 2: SET PATHS
# ======================================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Rainfall Intensity Classification"
)

DATASET_PATH = os.path.join(
    BASE_DIR,
    "rainfall_0.csv"
)

ACCURACY_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "accuracy_graph.png"
)

COMPARISON_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "comparison_graph.png"
)

HEATMAP_PATH = os.path.join(
    BASE_DIR,
    "heatmap.png"
)

RESULT_CSV_PATH = os.path.join(
    BASE_DIR,
    "result.csv"
)

RESULT_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "result_graph.png"
)

PREDICTION_CSV_PATH = os.path.join(
    BASE_DIR,
    "prediction.csv"
)

PREDICTION_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "prediction_graph.png"
)


os.makedirs(
    BASE_DIR,
    exist_ok=True
)


print("=" * 75)
print("AI-BASED RAINFALL INTENSITY CLASSIFICATION")
print("RESULT AND VISUALIZATION GENERATION")
print("=" * 75)


# ======================================================================
# STEP 3: RANDOM SEEDS
# ======================================================================

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)

tf.random.set_seed(
    RANDOM_STATE
)


# ======================================================================
# STEP 4: LOAD DATASET
# ======================================================================

if not os.path.exists(DATASET_PATH):

    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )


df = pd.read_csv(
    DATASET_PATH
)


print("\nDataset loaded successfully.")

print(
    "\nOriginal Dataset Shape:",
    df.shape
)


# ======================================================================
# STEP 5: CLEAN COLUMN NAMES
# ======================================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
)


print("\nDataset columns:")

for col in df.columns:

    print(col)


# ======================================================================
# STEP 6: DETECT LOCATION COLUMNS
# ======================================================================

possible_location_columns = [

    "District",
    "Taluka",
    "Taluk",
    "Circle",
    "Station",
    "Location",
    "Village"

]


location_columns = [

    col

    for col in df.columns

    if col.lower() in
    [x.lower() for x in possible_location_columns]

]


print(
    "\nDetected location columns:",
    location_columns
)


# ======================================================================
# STEP 7: DETECT RAINFALL COLUMNS
# ======================================================================

candidate_columns = [

    col

    for col in df.columns

    if col not in location_columns

]


rainfall_columns = []


for col in candidate_columns:

    numeric_data = pd.to_numeric(
        df[col],
        errors="coerce"
    )

    numeric_count = (
        numeric_data
        .notna()
        .sum()
    )

    minimum_required = max(
        3,
        int(len(df) * 0.30)
    )

    if numeric_count >= minimum_required:

        rainfall_columns.append(
            col
        )


if len(rainfall_columns) == 0:

    raise ValueError(
        "No rainfall columns detected."
    )


print(
    "\nNumber of rainfall columns:",
    len(rainfall_columns)
)


# ======================================================================
# STEP 8: CLEAN RAINFALL DATA
# ======================================================================

for col in rainfall_columns:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )


df[rainfall_columns] = (

    df[rainfall_columns]

    .replace(
        [np.inf, -np.inf],
        np.nan
    )

    .fillna(0)

)


df[rainfall_columns] = (
    df[rainfall_columns]
    .clip(lower=0)
)


print(
    "\nRainfall data cleaned successfully."
)


# ======================================================================
# STEP 9: FEATURE ENGINEERING
# ======================================================================

rainfall_data = (
    df[rainfall_columns]
)


df["Total_Rainfall"] = (
    rainfall_data
    .sum(axis=1)
)


df["Average_Rainfall"] = (
    rainfall_data
    .mean(axis=1)
)


df["Maximum_Rainfall"] = (
    rainfall_data
    .max(axis=1)
)


df["Minimum_Rainfall"] = (
    rainfall_data
    .min(axis=1)
)


df["Rainfall_Std"] = (
    rainfall_data
    .std(axis=1)
)


df["Rainy_Days"] = (

    rainfall_data > 0

).sum(axis=1)


df["Moderate_Rain_Days"] = (

    rainfall_data >= 15.6

).sum(axis=1)


df["Heavy_Rain_Days"] = (

    rainfall_data >= 64.5

).sum(axis=1)


df["Very_Heavy_Rain_Days"] = (

    rainfall_data >= 115.6

).sum(axis=1)


df["Extreme_Rain_Days"] = (

    rainfall_data >= 204.5

).sum(axis=1)


df["Rainfall_Range"] = (

    df["Maximum_Rainfall"]
    -
    df["Minimum_Rainfall"]

)


df["Rainfall_CV"] = (

    df["Rainfall_Std"]
    /
    (
        df["Average_Rainfall"]
        +
        1e-6
    )

)


print(
    "\nRainfall features generated successfully."
)


# ======================================================================
# STEP 10: AUGUST AND SEPTEMBER FEATURES
# ======================================================================

august_columns = [

    col

    for col in rainfall_columns

    if "aug" in col.lower()

]


september_columns = [

    col

    for col in rainfall_columns

    if "sep" in col.lower()

]


if len(august_columns) > 0:

    df["August_Total_Rainfall"] = (

        df[august_columns]
        .sum(axis=1)

    )

    df["August_Average_Rainfall"] = (

        df[august_columns]
        .mean(axis=1)

    )

else:

    df["August_Total_Rainfall"] = 0

    df["August_Average_Rainfall"] = 0


if len(september_columns) > 0:

    df["September_Total_Rainfall"] = (

        df[september_columns]
        .sum(axis=1)

    )

    df["September_Average_Rainfall"] = (

        df[september_columns]
        .mean(axis=1)

    )

else:

    df["September_Total_Rainfall"] = 0

    df["September_Average_Rainfall"] = 0


# ======================================================================
# STEP 11: CREATE TARGET CLASS
# ======================================================================

q1 = (
    df["Total_Rainfall"]
    .quantile(0.33)
)

q2 = (
    df["Total_Rainfall"]
    .quantile(0.66)
)


def classify_rainfall(value):

    if value <= q1:

        return "Low"

    elif value <= q2:

        return "Moderate"

    else:

        return "High"


df["Rainfall_Intensity"] = (

    df["Total_Rainfall"]
    .apply(
        classify_rainfall
    )

)


print(
    "\nRainfall classification thresholds:"
)

print(
    "Low <= ",
    round(q1, 2)
)

print(
    "Moderate <= ",
    round(q2, 2)
)

print(
    "High > ",
    round(q2, 2)
)


print(
    "\nClass Distribution:"
)

print(
    df["Rainfall_Intensity"]
    .value_counts()
)


# ======================================================================
# STEP 12: DEFINE FEATURES
# ======================================================================

feature_columns = [

    "Average_Rainfall",

    "Maximum_Rainfall",

    "Minimum_Rainfall",

    "Rainfall_Std",

    "Rainy_Days",

    "Moderate_Rain_Days",

    "Heavy_Rain_Days",

    "Very_Heavy_Rain_Days",

    "Extreme_Rain_Days",

    "Rainfall_Range",

    "Rainfall_CV",

    "August_Total_Rainfall",

    "August_Average_Rainfall",

    "September_Total_Rainfall",

    "September_Average_Rainfall"

]


X = (
    df[feature_columns]
    .copy()
)


X = (

    X
    .replace(
        [np.inf, -np.inf],
        np.nan
    )

    .fillna(0)

)


y = (
    df["Rainfall_Intensity"]
    .copy()
)


# ======================================================================
# STEP 13: LABEL ENCODING
# ======================================================================

label_encoder = (
    LabelEncoder()
)


y_encoded = (
    label_encoder
    .fit_transform(y)
)


print(
    "\nEncoded Classes:"
)


for index, label in enumerate(
    label_encoder.classes_
):

    print(
        index,
        "=",
        label
    )


# ======================================================================
# STEP 14: TRAIN TEST SPLIT
# ======================================================================

indices = np.arange(
    len(df)
)


(
    X_train,
    X_test,
    y_train,
    y_test,
    train_indices,
    test_indices

) = train_test_split(

    X,

    y_encoded,

    indices,

    test_size=0.25,

    random_state=RANDOM_STATE,

    stratify=y_encoded

)


print(
    "\nTraining Samples:",
    len(X_train)
)

print(
    "Testing Samples:",
    len(X_test)
)


# ======================================================================
# STEP 15: SCALE FEATURES
# ======================================================================

scaler = (
    StandardScaler()
)


X_train_scaled = (
    scaler
    .fit_transform(
        X_train
    )
)


X_test_scaled = (
    scaler
    .transform(
        X_test
    )
)


# ======================================================================
# STEP 16: RANDOM FOREST MODEL
# ======================================================================

print(
    "\nTraining Random Forest..."
)


rf_model = RandomForestClassifier(

    n_estimators=300,

    class_weight="balanced",

    random_state=RANDOM_STATE

)


rf_model.fit(

    X_train,

    y_train

)


rf_prediction = (
    rf_model
    .predict(
        X_test
    )
)


# ======================================================================
# STEP 17: RANDOM FOREST METRICS
# ======================================================================

rf_accuracy = accuracy_score(

    y_test,

    rf_prediction

)


rf_precision = precision_score(

    y_test,

    rf_prediction,

    average="weighted",

    zero_division=0

)


rf_recall = recall_score(

    y_test,

    rf_prediction,

    average="weighted",

    zero_division=0

)


rf_f1 = f1_score(

    y_test,

    rf_prediction,

    average="weighted",

    zero_division=0

)


print(
    "\nRandom Forest Accuracy:",
    round(
        rf_accuracy * 100,
        2
    ),
    "%"
)


# ======================================================================
# STEP 18: NEURAL NETWORK
# ======================================================================

print(
    "\nTraining Neural Network..."
)


number_of_classes = (
    len(
        label_encoder.classes_
    )
)


y_train_nn = (
    tf.keras.utils
    .to_categorical(

        y_train,

        num_classes=number_of_classes

    )
)


y_test_nn = (
    tf.keras.utils
    .to_categorical(

        y_test,

        num_classes=number_of_classes

    )
)


nn_model = Sequential([

    tf.keras.layers.Input(
        shape=(
            X_train_scaled.shape[1],
        )
    ),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(
        0.20
    ),

    Dense(
        32,
        activation="relu"
    ),

    Dropout(
        0.15
    ),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        number_of_classes,
        activation="softmax"
    )

])


nn_model.compile(

    optimizer="adam",

    loss="categorical_crossentropy",

    metrics=[
        "accuracy"
    ]

)


early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=20,

    restore_best_weights=True

)


history = nn_model.fit(

    X_train_scaled,

    y_train_nn,

    validation_split=0.20,

    epochs=200,

    batch_size=8,

    callbacks=[
        early_stopping
    ],

    verbose=1

)


# ======================================================================
# STEP 19: NEURAL NETWORK PREDICTIONS
# ======================================================================

nn_probability = (
    nn_model
    .predict(
        X_test_scaled,
        verbose=0
    )
)


nn_prediction = np.argmax(

    nn_probability,

    axis=1

)


nn_accuracy = accuracy_score(

    y_test,

    nn_prediction

)


nn_precision = precision_score(

    y_test,

    nn_prediction,

    average="weighted",

    zero_division=0

)


nn_recall = recall_score(

    y_test,

    nn_prediction,

    average="weighted",

    zero_division=0

)


nn_f1 = f1_score(

    y_test,

    nn_prediction,

    average="weighted",

    zero_division=0

)


print(
    "\nNeural Network Accuracy:",
    round(
        nn_accuracy * 100,
        2
    ),
    "%"
)


# ======================================================================
# STEP 20: ACCURACY GRAPH
# ======================================================================

print(
    "\nCreating accuracy_graph.png..."
)


epochs = range(
    1,
    len(
        history.history["accuracy"]
    )
    +
    1
)


plt.figure(
    figsize=(10, 6)
)


plt.plot(

    epochs,

    history.history["accuracy"],

    marker="o",

    markersize=3,

    label="Training Accuracy"

)


plt.plot(

    epochs,

    history.history["val_accuracy"],

    marker="s",

    markersize=3,

    label="Validation Accuracy"

)


plt.title(
    "Neural Network Training and Validation Accuracy",
    fontsize=14,
    fontweight="bold"
)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "Accuracy"
)


plt.ylim(
    0,
    1.05
)


plt.grid(
    alpha=0.3
)


plt.legend()


plt.tight_layout()


plt.savefig(

    ACCURACY_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 21: COMPARISON GRAPH
# ======================================================================

print(
    "\nCreating comparison_graph.png..."
)


comparison_data = pd.DataFrame({

    "Model": [

        "Random Forest",

        "Neural Network"

    ],

    "Accuracy": [

        rf_accuracy,

        nn_accuracy

    ],

    "Precision": [

        rf_precision,

        nn_precision

    ],

    "Recall": [

        rf_recall,

        nn_recall

    ],

    "F1 Score": [

        rf_f1,

        nn_f1

    ]

})


comparison_plot = (

    comparison_data
    .set_index(
        "Model"
    )

    * 100

)


ax = comparison_plot.plot(

    kind="bar",

    figsize=(11, 6),

    width=0.75

)


plt.title(

    "Machine Learning Model Performance Comparison",

    fontsize=14,

    fontweight="bold"

)


plt.xlabel(
    "Model"
)


plt.ylabel(
    "Score (%)"
)


plt.ylim(
    0,
    105
)


plt.xticks(
    rotation=0
)


plt.grid(
    axis="y",
    alpha=0.3
)


plt.legend(
    title="Metrics"
)


for container in ax.containers:

    ax.bar_label(

        container,

        fmt="%.1f",

        padding=3,

        fontsize=8

    )


plt.tight_layout()


plt.savefig(

    COMPARISON_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 22: CONFUSION MATRIX HEATMAP
# ======================================================================

print(
    "\nCreating heatmap.png..."
)


cm = confusion_matrix(

    y_test,

    rf_prediction

)


fig, ax = plt.subplots(

    figsize=(8, 6)

)


image = ax.imshow(

    cm,

    interpolation="nearest",

    cmap="Blues"

)


plt.colorbar(
    image
)


classes = (
    label_encoder.classes_
)


ax.set(

    xticks=np.arange(
        len(classes)
    ),

    yticks=np.arange(
        len(classes)
    ),

    xticklabels=classes,

    yticklabels=classes,

    xlabel="Predicted Rainfall Intensity",

    ylabel="Actual Rainfall Intensity",

    title="Random Forest Confusion Matrix"

)


threshold = (
    cm.max() / 2
    if cm.size > 0
    else 0
)


for i in range(
    cm.shape[0]
):

    for j in range(
        cm.shape[1]
    ):

        ax.text(

            j,

            i,

            str(
                cm[i, j]
            ),

            ha="center",

            va="center",

            color=(
                "white"
                if cm[i, j] > threshold
                else "black"
            ),

            fontsize=12,

            fontweight="bold"

        )


plt.tight_layout()


plt.savefig(

    HEATMAP_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 23: CREATE RESULT CSV
# ======================================================================

print(
    "\nCreating result.csv..."
)


actual_labels = (
    label_encoder
    .inverse_transform(
        y_test
    )
)


rf_labels = (
    label_encoder
    .inverse_transform(
        rf_prediction
    )
)


nn_labels = (
    label_encoder
    .inverse_transform(
        nn_prediction
    )
)


result_df = (

    df.iloc[test_indices]
    .copy()
    .reset_index(drop=True)

)


result_output = pd.DataFrame()


# Add location information
for col in location_columns:

    result_output[col] = (
        result_df[col]
    )


result_output[
    "Total_Rainfall"
] = (
    result_df[
        "Total_Rainfall"
    ]
)


result_output[
    "Average_Rainfall"
] = (
    result_df[
        "Average_Rainfall"
    ]
)


result_output[
    "Maximum_Rainfall"
] = (
    result_df[
        "Maximum_Rainfall"
    ]
)


result_output[
    "Rainy_Days"
] = (
    result_df[
        "Rainy_Days"
    ]
)


result_output[
    "Heavy_Rain_Days"
] = (
    result_df[
        "Heavy_Rain_Days"
    ]
)


result_output[
    "Actual_Intensity"
] = (
    actual_labels
)


result_output[
    "Random_Forest_Prediction"
] = (
    rf_labels
)


result_output[
    "Neural_Network_Prediction"
] = (
    nn_labels
)


result_output[
    "RF_Correct"
] = (

    result_output[
        "Actual_Intensity"
    ]

    ==

    result_output[
        "Random_Forest_Prediction"
    ]

)


result_output[
    "NN_Correct"
] = (

    result_output[
        "Actual_Intensity"
    ]

    ==

    result_output[
        "Neural_Network_Prediction"
    ]

)


result_output.to_csv(

    RESULT_CSV_PATH,

    index=False

)


print(
    "\nresult.csv saved:"
)

print(
    RESULT_CSV_PATH
)


print(
    "\nResult Preview:"
)

print(
    result_output.head(10)
)


# ======================================================================
# STEP 24: RESULT GRAPH
# ======================================================================

print(
    "\nCreating result_graph.png..."
)


actual_counts = (

    result_output[
        "Actual_Intensity"
    ]

    .value_counts()

    .reindex(
        classes,
        fill_value=0
    )

)


rf_counts = (

    result_output[
        "Random_Forest_Prediction"
    ]

    .value_counts()

    .reindex(
        classes,
        fill_value=0
    )

)


nn_counts = (

    result_output[
        "Neural_Network_Prediction"
    ]

    .value_counts()

    .reindex(
        classes,
        fill_value=0
    )

)


x = np.arange(
    len(classes)
)


width = 0.25


plt.figure(
    figsize=(10, 6)
)


plt.bar(

    x - width,

    actual_counts.values,

    width,

    label="Actual"

)


plt.bar(

    x,

    rf_counts.values,

    width,

    label="Random Forest"

)


plt.bar(

    x + width,

    nn_counts.values,

    width,

    label="Neural Network"

)


plt.title(

    "Actual vs Predicted Rainfall Intensity",

    fontsize=14,

    fontweight="bold"

)


plt.xlabel(
    "Rainfall Intensity"
)


plt.ylabel(
    "Number of Records"
)


plt.xticks(

    x,

    classes

)


plt.legend()


plt.grid(

    axis="y",

    alpha=0.3

)


plt.tight_layout()


plt.savefig(

    RESULT_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 25: GENERATE PREDICTIONS FOR COMPLETE DATASET
# ======================================================================

print(
    "\nGenerating predictions for complete dataset..."
)


X_all = (

    df[feature_columns]
    .copy()
)


X_all = (

    X_all

    .replace(
        [np.inf, -np.inf],
        np.nan
    )

    .fillna(0)

)


# Random Forest prediction
rf_all_prediction = (

    rf_model
    .predict(
        X_all
    )

)


rf_all_labels = (

    label_encoder
    .inverse_transform(
        rf_all_prediction
    )

)


# Neural Network prediction
X_all_scaled = (

    scaler
    .transform(
        X_all
    )

)


nn_all_probability = (

    nn_model
    .predict(

        X_all_scaled,

        verbose=0

    )

)


nn_all_prediction = np.argmax(

    nn_all_probability,

    axis=1

)


nn_all_labels = (

    label_encoder
    .inverse_transform(
        nn_all_prediction
    )

)


# ======================================================================
# STEP 26: CREATE PREDICTION CSV
# ======================================================================

prediction_output = pd.DataFrame()


for col in location_columns:

    prediction_output[col] = (
        df[col]
    )


prediction_output[
    "Total_Rainfall"
] = (
    df[
        "Total_Rainfall"
    ]
)


prediction_output[
    "Average_Rainfall"
] = (
    df[
        "Average_Rainfall"
    ]
)


prediction_output[
    "Maximum_Rainfall"
] = (
    df[
        "Maximum_Rainfall"
    ]
)


prediction_output[
    "Rainfall_Std"
] = (
    df[
        "Rainfall_Std"
    ]
)


prediction_output[
    "Rainy_Days"
] = (
    df[
        "Rainy_Days"
    ]
)


prediction_output[
    "Heavy_Rain_Days"
] = (
    df[
        "Heavy_Rain_Days"
    ]
)


prediction_output[
    "Actual_Intensity"
] = (
    df[
        "Rainfall_Intensity"
    ]
)


prediction_output[
    "Random_Forest_Prediction"
] = (
    rf_all_labels
)


prediction_output[
    "Neural_Network_Prediction"
] = (
    nn_all_labels
)


# Neural Network confidence
prediction_output[
    "NN_Confidence"
] = (

    np.max(
        nn_all_probability,
        axis=1
    )

    * 100

)


# Random Forest confidence
rf_all_probability = (

    rf_model
    .predict_proba(
        X_all
    )

)


prediction_output[
    "RF_Confidence"
] = (

    np.max(
        rf_all_probability,
        axis=1
    )

    * 100

)


prediction_output.to_csv(

    PREDICTION_CSV_PATH,

    index=False

)


print(
    "\nprediction.csv saved:"
)

print(
    PREDICTION_CSV_PATH
)


print(
    "\nPrediction Preview:"
)

print(
    prediction_output.head(10)
)


# ======================================================================
# STEP 27: PREDICTION GRAPH
# ======================================================================

print(
    "\nCreating prediction_graph.png..."
)


prediction_counts = (

    prediction_output[
        "Random_Forest_Prediction"
    ]

    .value_counts()

    .reindex(
        classes,
        fill_value=0
    )

)


plt.figure(
    figsize=(9, 6)
)


bars = plt.bar(

    prediction_counts.index,

    prediction_counts.values

)


plt.title(

    "Predicted Rainfall Intensity Distribution",

    fontsize=14,

    fontweight="bold"

)


plt.xlabel(
    "Rainfall Intensity Class"
)


plt.ylabel(
    "Number of Locations"
)


plt.grid(

    axis="y",

    alpha=0.3

)


for bar in bars:

    height = (
        bar.get_height()
    )

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        height,

        str(
            int(height)
        ),

        ha="center",

        va="bottom",

        fontweight="bold"

    )


plt.tight_layout()


plt.savefig(

    PREDICTION_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 28: FEATURE IMPORTANCE GRAPH
# ======================================================================

feature_importance = pd.DataFrame({

    "Feature":
        feature_columns,

    "Importance":
        rf_model.feature_importances_

})


feature_importance = (

    feature_importance

    .sort_values(

        "Importance",

        ascending=True

    )

)


FEATURE_IMPORTANCE_PATH = os.path.join(

    BASE_DIR,

    "feature_importance_graph.png"

)


plt.figure(
    figsize=(10, 8)
)


plt.barh(

    feature_importance[
        "Feature"
    ],

    feature_importance[
        "Importance"
    ]

)


plt.title(

    "Rainfall Feature Importance",

    fontsize=14,

    fontweight="bold"

)


plt.xlabel(
    "Importance Score"
)


plt.ylabel(
    "Feature"
)


plt.grid(

    axis="x",

    alpha=0.3

)


plt.tight_layout()


plt.savefig(

    FEATURE_IMPORTANCE_PATH,

    dpi=300,

    bbox_inches="tight"

)


plt.show()

plt.close()


# ======================================================================
# STEP 29: PRINT CLASSIFICATION REPORTS
# ======================================================================

print(
    "\n"
    +
    "=" * 75
)

print(
    "RANDOM FOREST CLASSIFICATION REPORT"
)

print(
    "=" * 75
)


print(

    classification_report(

        y_test,

        rf_prediction,

        target_names=classes,

        zero_division=0

    )

)


print(
    "\n"
    +
    "=" * 75
)

print(
    "NEURAL NETWORK CLASSIFICATION REPORT"
)

print(
    "=" * 75
)


print(

    classification_report(

        y_test,

        nn_prediction,

        target_names=classes,

        zero_division=0

    )

)


# ======================================================================
# STEP 30: PRINT MODEL COMPARISON
# ======================================================================

print(
    "\nMODEL COMPARISON"
)


print(
    comparison_data.to_string(
        index=False
    )
)


# ======================================================================
# STEP 31: FINAL OUTPUT SUMMARY
# ======================================================================

print(
    "\n"
    +
    "=" * 75
)

print(
    "ALL OUTPUTS GENERATED SUCCESSFULLY"
)

print(
    "=" * 75
)


print(
    "\nRandom Forest Accuracy:"
)

print(
    f"{rf_accuracy * 100:.2f}%"
)


print(
    "\nNeural Network Accuracy:"
)

print(
    f"{nn_accuracy * 100:.2f}%"
)


print(
    "\nGenerated Files:"
)


generated_files = [

    ACCURACY_GRAPH_PATH,

    COMPARISON_GRAPH_PATH,

    HEATMAP_PATH,

    RESULT_CSV_PATH,

    RESULT_GRAPH_PATH,

    PREDICTION_CSV_PATH,

    PREDICTION_GRAPH_PATH,

    FEATURE_IMPORTANCE_PATH

]


for number, file_path in enumerate(

    generated_files,

    start=1

):

    print(
        f"{number}. {file_path}"
    )


print(
    "\nProject visualization and result generation completed."
)
```
