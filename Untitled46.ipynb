```python
import os
import json
import warnings
import joblib
import yaml
import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping


# ======================================================================
# STEP 1: SET PATHS
# ======================================================================

BASE_DIR = r"C:\Users\sagni\Downloads\AI-Based Rainfall Intensity Classification"

DATASET_PATH = os.path.join(
    BASE_DIR,
    "rainfall_0.csv"
)

H5_PATH = os.path.join(
    BASE_DIR,
    "rainfall_model.h5"
)

PKL_PATH = os.path.join(
    BASE_DIR,
    "rainfall_model.pkl"
)

YAML_PATH = os.path.join(
    BASE_DIR,
    "rainfall_config.yaml"
)

JSON_PATH = os.path.join(
    BASE_DIR,
    "rainfall_metadata.json"
)

os.makedirs(BASE_DIR, exist_ok=True)

print("=" * 70)
print("AI-BASED RAINFALL INTENSITY CLASSIFICATION")
print("=" * 70)

print("\nDataset path:")
print(DATASET_PATH)

print("\nOutput directory:")
print(BASE_DIR)


# ======================================================================
# STEP 2: SET RANDOM SEEDS
# ======================================================================

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
tf.random.set_seed(RANDOM_STATE)


# ======================================================================
# STEP 3: LOAD DATASET
# ======================================================================

if not os.path.exists(DATASET_PATH):
    raise FileNotFoundError(
        f"\nDataset not found:\n{DATASET_PATH}\n"
        "Please check the file path."
    )

df = pd.read_csv(DATASET_PATH)

print("\nDataset loaded successfully!")

print("\nDataset Shape:")
print(df.shape)

print("\nFirst 5 rows:")
print(df.head())

print("\nColumns:")
for col in df.columns:
    print(col)


# ======================================================================
# STEP 4: CLEAN COLUMN NAMES
# ======================================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
)

print("\nCleaned column names:")
print(df.columns.tolist())


# ======================================================================
# STEP 5: IDENTIFY LOCATION AND RAINFALL COLUMNS
# ======================================================================

# Common location columns in rainfall datasets
possible_location_columns = [
    "District",
    "Taluka",
    "Taluk",
    "Circle",
    "Station",
    "Location",
    "Village"
]

location_columns = [
    col for col in df.columns
    if col.strip().lower()
    in [x.lower() for x in possible_location_columns]
]

print("\nDetected location columns:")
print(location_columns)


# ======================================================================
# STEP 6: IDENTIFY DAILY RAINFALL COLUMNS
# ======================================================================

# Everything other than location columns is initially considered
# a possible rainfall column.

candidate_columns = [
    col for col in df.columns
    if col not in location_columns
]

rainfall_columns = []

for col in candidate_columns:

    converted = pd.to_numeric(
        df[col],
        errors="coerce"
    )

    # Keep columns that contain enough numeric rainfall observations
    if converted.notna().sum() >= max(3, int(len(df) * 0.30)):
        rainfall_columns.append(col)


if len(rainfall_columns) == 0:
    raise ValueError(
        "No rainfall measurement columns could be detected."
    )

print("\nDetected rainfall columns:")

for col in rainfall_columns:
    print(col)

print(
    "\nNumber of rainfall measurement columns:",
    len(rainfall_columns)
)


# ======================================================================
# STEP 7: CONVERT RAINFALL VALUES TO NUMERIC
# ======================================================================

for col in rainfall_columns:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

# Rainfall missing values are assumed to represent zero rainfall
df[rainfall_columns] = (
    df[rainfall_columns]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)

# Rainfall cannot normally be negative
df[rainfall_columns] = df[rainfall_columns].clip(lower=0)

print("\nRainfall columns converted to numeric successfully.")


# ======================================================================
# STEP 8: CREATE RAINFALL FEATURES
# ======================================================================

rainfall_data = df[rainfall_columns]

df["Total_Rainfall"] = rainfall_data.sum(axis=1)

df["Average_Rainfall"] = rainfall_data.mean(axis=1)

df["Maximum_Rainfall"] = rainfall_data.max(axis=1)

df["Minimum_Rainfall"] = rainfall_data.min(axis=1)

df["Rainfall_Std"] = rainfall_data.std(axis=1)

df["Rainy_Days"] = (
    rainfall_data > 0
).sum(axis=1)

df["Moderate_Rain_Days"] = (
    rainfall_data >= 15.6
).sum(axis=1)

df["Heavy_Rain_Days"] = (
    rainfall_data >= 64.5
).sum(axis=1)

df["Very_Heavy_Rain_Days"] = (
    rainfall_data >= 115.6
).sum(axis=1)

df["Extreme_Rain_Days"] = (
    rainfall_data >= 204.5
).sum(axis=1)


# ======================================================================
# STEP 9: CREATE RAINFALL CONSISTENCY FEATURES
# ======================================================================

df["Rainfall_Range"] = (
    df["Maximum_Rainfall"]
    -
    df["Minimum_Rainfall"]
)

df["Rainfall_CV"] = (
    df["Rainfall_Std"]
    /
    (df["Average_Rainfall"] + 1e-6)
)

print("\nRainfall features created successfully.")


# ======================================================================
# STEP 10: OPTIONAL MONTH-WISE FEATURES
# ======================================================================

august_columns = [
    col for col in rainfall_columns
    if "aug" in col.lower()
]

september_columns = [
    col for col in rainfall_columns
    if "sep" in col.lower()
]


if len(august_columns) > 0:

    df["August_Total_Rainfall"] = (
        df[august_columns].sum(axis=1)
    )

    df["August_Average_Rainfall"] = (
        df[august_columns].mean(axis=1)
    )

else:

    df["August_Total_Rainfall"] = 0
    df["August_Average_Rainfall"] = 0


if len(september_columns) > 0:

    df["September_Total_Rainfall"] = (
        df[september_columns].sum(axis=1)
    )

    df["September_Average_Rainfall"] = (
        df[september_columns].mean(axis=1)
    )

else:

    df["September_Total_Rainfall"] = 0
    df["September_Average_Rainfall"] = 0


# ======================================================================
# STEP 11: CREATE TARGET CLASS
# ======================================================================

# The dataset does not necessarily contain an existing class label.
# Therefore rainfall intensity classes are generated from the distribution
# of Total_Rainfall.
#
# qcut attempts to create approximately balanced:
# Low / Moderate / High rainfall groups.

try:

    df["Rainfall_Intensity"] = pd.qcut(
        df["Total_Rainfall"],
        q=3,
        labels=[
            "Low",
            "Moderate",
            "High"
        ],
        duplicates="drop"
    )

except Exception:

    q1 = df["Total_Rainfall"].quantile(0.33)
    q2 = df["Total_Rainfall"].quantile(0.66)

    def rainfall_class(value):

        if value <= q1:
            return "Low"

        elif value <= q2:
            return "Moderate"

        else:
            return "High"

    df["Rainfall_Intensity"] = (
        df["Total_Rainfall"]
        .apply(rainfall_class)
    )


# Check whether qcut successfully produced three classes
if df["Rainfall_Intensity"].nunique() < 3:

    q1 = df["Total_Rainfall"].quantile(0.33)
    q2 = df["Total_Rainfall"].quantile(0.66)

    def rainfall_class_fallback(value):

        if value <= q1:
            return "Low"

        elif value <= q2:
            return "Moderate"

        return "High"

    df["Rainfall_Intensity"] = (
        df["Total_Rainfall"]
        .apply(rainfall_class_fallback)
    )


df["Rainfall_Intensity"] = (
    df["Rainfall_Intensity"]
    .astype(str)
)


print("\nRainfall Intensity Distribution:")

print(
    df["Rainfall_Intensity"]
    .value_counts()
)


# ======================================================================
# STEP 12: DEFINE MACHINE LEARNING FEATURES
# ======================================================================

feature_columns = [

    "Average_Rainfall",
    "Maximum_Rainfall",
    "Minimum_Rainfall",
    "Rainfall_Std",

    "Rainy_Days",
    "Moderate_Rain_Days",
    "Heavy_Rain_Days",
    "Very_Heavy_Rain_Days",
    "Extreme_Rain_Days",

    "Rainfall_Range",
    "Rainfall_CV",

    "August_Total_Rainfall",
    "August_Average_Rainfall",

    "September_Total_Rainfall",
    "September_Average_Rainfall"
]


X = df[feature_columns].copy()

y = df["Rainfall_Intensity"].copy()


print("\nFeatures used:")

for feature in feature_columns:
    print(feature)


# ======================================================================
# STEP 13: HANDLE INVALID VALUES
# ======================================================================

X = X.replace(
    [np.inf, -np.inf],
    np.nan
)

X = X.fillna(0)


# ======================================================================
# STEP 14: ENCODE TARGET
# ======================================================================

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("\nTarget classes:")

for i, label in enumerate(label_encoder.classes_):
    print(i, "=", label)


# ======================================================================
# STEP 15: TRAIN TEST SPLIT
# ======================================================================

# Stratification is used where possible.

try:

    X_train, X_test, y_train, y_test = train_test_split(

        X,
        y_encoded,

        test_size=0.25,

        random_state=RANDOM_STATE,

        stratify=y_encoded
    )

except ValueError:

    X_train, X_test, y_train, y_test = train_test_split(

        X,
        y_encoded,

        test_size=0.25,

        random_state=RANDOM_STATE
    )


print("\nTraining samples:", len(X_train))
print("Testing samples :", len(X_test))


# ======================================================================
# STEP 16: FEATURE SCALING
# ======================================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)


# ======================================================================
# STEP 17: RANDOM FOREST MODEL
# ======================================================================

print("\n" + "=" * 70)
print("TRAINING RANDOM FOREST MODEL")
print("=" * 70)


rf_model = RandomForestClassifier(

    n_estimators=300,

    max_depth=None,

    min_samples_split=2,

    min_samples_leaf=1,

    class_weight="balanced",

    random_state=RANDOM_STATE
)


rf_model.fit(
    X_train,
    y_train
)


# ======================================================================
# STEP 18: RANDOM FOREST PREDICTION
# ======================================================================

rf_predictions = rf_model.predict(
    X_test
)

rf_accuracy = accuracy_score(
    y_test,
    rf_predictions
)

print(
    "\nRandom Forest Accuracy:",
    round(rf_accuracy * 100, 2),
    "%"
)


print("\nClassification Report:")

print(
    classification_report(
        y_test,
        rf_predictions,
        target_names=label_encoder.classes_,
        zero_division=0
    )
)


print("\nConfusion Matrix:")

print(
    confusion_matrix(
        y_test,
        rf_predictions
    )
)


# ======================================================================
# STEP 19: SAVE RANDOM FOREST MODEL AS PKL
# ======================================================================

model_package = {

    "model": rf_model,

    "scaler": scaler,

    "label_encoder": label_encoder,

    "feature_columns": feature_columns,

    "rainfall_columns": rainfall_columns
}


joblib.dump(
    model_package,
    PKL_PATH
)


print("\nPKL model saved successfully:")
print(PKL_PATH)


# ======================================================================
# STEP 20: PREPARE DATA FOR NEURAL NETWORK
# ======================================================================

num_classes = len(
    label_encoder.classes_
)

y_train_nn = tf.keras.utils.to_categorical(
    y_train,
    num_classes=num_classes
)

y_test_nn = tf.keras.utils.to_categorical(
    y_test,
    num_classes=num_classes
)


# ======================================================================
# STEP 21: BUILD NEURAL NETWORK
# ======================================================================

print("\n" + "=" * 70)
print("TRAINING NEURAL NETWORK")
print("=" * 70)


nn_model = Sequential([

    tf.keras.layers.Input(
        shape=(X_train_scaled.shape[1],)
    ),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(0.20),

    Dense(
        32,
        activation="relu"
    ),

    Dropout(0.15),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        num_classes,
        activation="softmax"
    )

])


# ======================================================================
# STEP 22: COMPILE NEURAL NETWORK
# ======================================================================

nn_model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="categorical_crossentropy",

    metrics=["accuracy"]
)


print("\nNeural Network Architecture:")

nn_model.summary()


# ======================================================================
# STEP 23: EARLY STOPPING
# ======================================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=20,

    restore_best_weights=True,

    verbose=1
)


# ======================================================================
# STEP 24: TRAIN NEURAL NETWORK
# ======================================================================

history = nn_model.fit(

    X_train_scaled,

    y_train_nn,

    validation_split=0.20,

    epochs=200,

    batch_size=8,

    callbacks=[
        early_stopping
    ],

    verbose=1
)


# ======================================================================
# STEP 25: EVALUATE NEURAL NETWORK
# ======================================================================

nn_loss, nn_accuracy = nn_model.evaluate(

    X_test_scaled,

    y_test_nn,

    verbose=0
)


print(
    "\nNeural Network Accuracy:",
    round(nn_accuracy * 100, 2),
    "%"
)


# ======================================================================
# STEP 26: NEURAL NETWORK PREDICTIONS
# ======================================================================

nn_probabilities = nn_model.predict(
    X_test_scaled,
    verbose=0
)

nn_predictions = np.argmax(
    nn_probabilities,
    axis=1
)


print("\nNeural Network Classification Report:")

print(
    classification_report(
        y_test,
        nn_predictions,
        target_names=label_encoder.classes_,
        zero_division=0
    )
)


# ======================================================================
# STEP 27: SAVE NEURAL NETWORK AS H5
# ======================================================================

nn_model.save(
    H5_PATH
)


print("\nH5 model saved successfully:")
print(H5_PATH)


# ======================================================================
# STEP 28: CREATE YAML CONFIGURATION
# ======================================================================

yaml_config = {

    "project": {
        "name":
            "AI-Based Rainfall Intensity Classification",

        "description":
            "Machine Learning and Neural Network based rainfall "
            "intensity classification system."
    },

    "dataset": {
        "filename":
            os.path.basename(DATASET_PATH),

        "rows":
            int(df.shape[0]),

        "columns":
            int(df.shape[1]),

        "rainfall_measurement_columns":
            int(len(rainfall_columns))
    },

    "target": {
        "column":
            "Rainfall_Intensity",

        "classes":
            label_encoder.classes_.tolist()
    },

    "features":
        feature_columns,

    "models": {

        "random_forest": {

            "type":
                "RandomForestClassifier",

            "estimators":
                300,

            "accuracy":
                float(rf_accuracy)
        },

        "neural_network": {

            "type":
                "Feedforward Neural Network",

            "accuracy":
                float(nn_accuracy),

            "epochs_requested":
                200
        }
    },

    "saved_files": {

        "h5":
            os.path.basename(H5_PATH),

        "pkl":
            os.path.basename(PKL_PATH),

        "yaml":
            os.path.basename(YAML_PATH),

        "json":
            os.path.basename(JSON_PATH)
    }
}


with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        yaml_config,
        file,
        sort_keys=False,
        allow_unicode=True
    )


print("\nYAML configuration saved successfully:")
print(YAML_PATH)


# ======================================================================
# STEP 29: CREATE JSON METADATA
# ======================================================================

metadata = {

    "project_name":
        "AI-Based Rainfall Intensity Classification",

    "dataset":
        os.path.basename(DATASET_PATH),

    "dataset_shape": {

        "rows":
            int(df.shape[0]),

        "columns":
            int(df.shape[1])
    },

    "location_columns":
        location_columns,

    "rainfall_columns":
        rainfall_columns,

    "feature_columns":
        feature_columns,

    "target_column":
        "Rainfall_Intensity",

    "target_classes":
        label_encoder.classes_.tolist(),

    "class_distribution":
        {
            str(key): int(value)

            for key, value in
            df["Rainfall_Intensity"]
            .value_counts()
            .to_dict()
            .items()
        },

    "random_forest_accuracy":
        float(rf_accuracy),

    "neural_network_accuracy":
        float(nn_accuracy),

    "training_samples":
        int(len(X_train)),

    "testing_samples":
        int(len(X_test)),

    "random_state":
        RANDOM_STATE
}


with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4
    )


print("\nJSON metadata saved successfully:")
print(JSON_PATH)


# ======================================================================
# STEP 30: TEST LOADING PKL MODEL
# ======================================================================

print("\n" + "=" * 70)
print("TESTING SAVED PKL MODEL")
print("=" * 70)


loaded_package = joblib.load(
    PKL_PATH
)

loaded_rf_model = loaded_package["model"]

sample = X.iloc[[0]]

sample_prediction = loaded_rf_model.predict(
    sample
)

sample_label = label_encoder.inverse_transform(
    sample_prediction
)[0]


print("\nSample Random Forest Prediction:")
print(sample_label)


# ======================================================================
# STEP 31: TEST LOADING H5 MODEL
# ======================================================================

print("\n" + "=" * 70)
print("TESTING SAVED H5 MODEL")
print("=" * 70)


loaded_nn_model = tf.keras.models.load_model(
    H5_PATH
)

sample_scaled = scaler.transform(
    sample
)

sample_nn_probability = loaded_nn_model.predict(
    sample_scaled,
    verbose=0
)

sample_nn_index = np.argmax(
    sample_nn_probability,
    axis=1
)

sample_nn_label = label_encoder.inverse_transform(
    sample_nn_index
)[0]


print("\nSample Neural Network Prediction:")
print(sample_nn_label)


# ======================================================================
# STEP 32: DISPLAY FEATURE IMPORTANCE
# ======================================================================

feature_importance = pd.DataFrame({

    "Feature":
        feature_columns,

    "Importance":
        rf_model.feature_importances_

})


feature_importance = feature_importance.sort_values(

    "Importance",

    ascending=False

).reset_index(drop=True)


print("\nRandom Forest Feature Importance:")

print(feature_importance)


# ======================================================================
# STEP 33: DISPLAY SAMPLE RESULTS
# ======================================================================

display_columns = []

for col in location_columns:
    display_columns.append(col)

display_columns.extend([

    "Total_Rainfall",
    "Average_Rainfall",
    "Maximum_Rainfall",
    "Rainy_Days",
    "Heavy_Rain_Days",
    "Rainfall_Intensity"

])


print("\nSample Rainfall Classification:")

print(
    df[display_columns]
    .head(10)
    .to_string(index=False)
)


# ======================================================================
# STEP 34: FINAL SUMMARY
# ======================================================================

print("\n" + "=" * 70)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 70)


print("\nProject:")
print("AI-Based Rainfall Intensity Classification")


print("\nRandom Forest Accuracy:")
print(
    f"{rf_accuracy * 100:.2f}%"
)


print("\nNeural Network Accuracy:")
print(
    f"{nn_accuracy * 100:.2f}%"
)


print("\nGenerated Files:")

print(
    "1.",
    H5_PATH
)

print(
    "2.",
    PKL_PATH
)

print(
    "3.",
    YAML_PATH
)

print(
    "4.",
    JSON_PATH
)


print("\nAll required model files have been generated successfully.")
```
